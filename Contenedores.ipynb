# Introducción a la Contenerización

En los proyectos modernos de software y datos es común que una misma aplicación deba ejecutarse en distintos entornos, como computadoras de desarrollo, servidores de pruebas, ambientes de integración continua y servidores de producción. Aunque el código fuente sea exactamente el mismo, pequeñas diferencias en configuraciones, sistemas operativos o versiones de software pueden provocar errores inesperados.

La contenerización surge como una solución para garantizar que una aplicación funcione de manera consistente independientemente del entorno donde se ejecute. Esta tecnología se ha convertido en uno de los pilares fundamentales del desarrollo moderno, la computación en la nube y la ingeniería de datos.

## Problemas de Compatibilidad entre Entornos

Uno de los problemas más frecuentes en el desarrollo de software ocurre cuando una aplicación funciona correctamente en el entorno del desarrollador, pero presenta errores al ser desplegada en otro sistema.

Las causas más comunes incluyen:

- Diferencias en versiones del sistema operativo.
- Dependencias instaladas incorrectamente.
- Versiones distintas de lenguajes de programación.
- Configuraciones incompatibles.
- Variables de entorno faltantes.
- Conflictos entre aplicaciones instaladas en el mismo servidor.

Por ejemplo, un desarrollador puede crear una aplicación utilizando Python 3.12 y una versión específica de una librería. Si el servidor utiliza Python 3.10 o una versión diferente de dicha librería, la aplicación podría comportarse de manera distinta o incluso fallar durante su ejecución.

Este tipo de situaciones generan retrasos, errores de despliegue y dificultades para mantener ambientes consistentes entre los miembros de un equipo de trabajo.

## Evolución de la Virtualización a los Contenedores

Antes de la aparición de los contenedores, la principal estrategia para aislar aplicaciones consistía en utilizar máquinas virtuales.

Una máquina virtual permite ejecutar un sistema operativo completo dentro de otro sistema operativo mediante un software conocido como hipervisor.

La estructura típica de una solución basada en máquinas virtuales puede representarse de la siguiente manera:

```text
Servidor Físico
│
├── Hipervisor
│
├── Máquina Virtual 1
│   ├── Sistema Operativo
│   └── Aplicación
│
├── Máquina Virtual 2
│   ├── Sistema Operativo
│   └── Aplicación
│
└── Máquina Virtual 3
    ├── Sistema Operativo
    └── Aplicación
```

Las máquinas virtuales ofrecen un alto nivel de aislamiento, pero también presentan algunas limitaciones:

- Mayor consumo de memoria.
- Mayor uso de almacenamiento.
- Tiempos de inicio más lentos.
- Necesidad de administrar múltiples sistemas operativos.

A medida que las aplicaciones crecieron en complejidad, surgió la necesidad de una solución más ligera y eficiente. Esto llevó al desarrollo de los contenedores.

## ¿Qué es la Contenerización?

La contenerización es una técnica que permite empaquetar una aplicación junto con todos los elementos necesarios para su funcionamiento.

Un contenedor puede incluir:

- Código fuente.
- Dependencias.
- Librerías.
- Configuraciones.
- Variables de entorno.
- Herramientas necesarias para la ejecución.

Gracias a este enfoque, la aplicación puede ejecutarse de forma consistente en cualquier entorno que disponga de un motor de contenedores compatible.

En lugar de depender de la configuración del sistema donde se despliega, el contenedor lleva consigo todo lo necesario para funcionar correctamente.

## Diferencia Conceptual entre Máquinas Virtuales y Contenedores

Mientras que una máquina virtual incluye un sistema operativo completo para cada instancia, los contenedores comparten el sistema operativo anfitrión.

La arquitectura simplificada de contenedores puede representarse de la siguiente forma:

```text
Servidor Físico
│
├── Sistema Operativo Host
│
├── Motor de Contenedores
│
├── Contenedor 1
│   └── Aplicación
│
├── Contenedor 2
│   └── Aplicación
│
└── Contenedor 3
    └── Aplicación
```

Esta diferencia permite que los contenedores sean significativamente más ligeros y rápidos que las máquinas virtuales.

## Beneficios de la Contenerización

La adopción de contenedores ha crecido rápidamente debido a las ventajas que ofrecen tanto a desarrolladores como a equipos de operaciones.

### Portabilidad

Los contenedores pueden ejecutarse en distintos entornos sin necesidad de realizar modificaciones en la aplicación.

### Consistencia

Garantizan que una aplicación funcione de la misma manera en desarrollo, pruebas y producción.

### Rapidez

Los tiempos de inicio son considerablemente menores en comparación con las máquinas virtuales.

### Escalabilidad

Permiten desplegar múltiples instancias de una aplicación de manera rápida y eficiente.

### Aislamiento

Cada contenedor funciona de manera independiente, evitando conflictos entre aplicaciones.

### Eficiencia de Recursos

Al compartir el sistema operativo anfitrión, consumen menos memoria y almacenamiento.

### Facilidad de Distribución

Los contenedores pueden compartirse fácilmente mediante repositorios especializados, facilitando el trabajo colaborativo y los procesos de despliegue.

## Casos de Uso en Ingeniería de Datos

La contenerización se ha convertido en una herramienta esencial dentro de la ingeniería de datos debido a la gran cantidad de tecnologías que normalmente participan en una solución moderna.

Algunos casos de uso comunes incluyen:

### Ambientes de Desarrollo Reproducibles

Permiten que todos los integrantes de un equipo trabajen con exactamente las mismas versiones de herramientas y dependencias.

Ejemplos:

- Python
- PostgreSQL
- Apache Airflow
- Jupyter Notebook
- Apache Spark

### Procesos ETL

Los procesos de extracción, transformación y carga pueden ejecutarse dentro de contenedores independientes, simplificando su despliegue y mantenimiento.

### Plataformas de Orquestación

Herramientas como Airflow pueden desplegarse mediante contenedores para facilitar la administración de sus componentes.

### Laboratorios de Aprendizaje

Los estudiantes pueden crear entornos completos de práctica sin necesidad de instalar manualmente cada herramienta.

### Integración Continua y Despliegue Continuo

Los pipelines de automatización utilizan contenedores para ejecutar pruebas, validaciones y despliegues de manera uniforme.

## Consideraciones Finales

La contenerización no reemplaza completamente a las máquinas virtuales. Ambas tecnologías continúan siendo utilizadas en la industria según las necesidades de cada proyecto.

En términos generales:

- Las máquinas virtuales ofrecen un mayor nivel de aislamiento.
- Los contenedores ofrecen mayor eficiencia y velocidad.
- Ambas tecnologías pueden coexistir dentro de una misma arquitectura.

Actualmente, la contenerización es una habilidad fundamental para cualquier profesional relacionado con desarrollo de software, ingeniería de datos, DevOps o computación en la nube, ya que permite construir soluciones portables, escalables y reproducibles con menor esfuerzo operativo.

# Conceptos Fundamentales de Contenedores

Los contenedores son una de las tecnologías más importantes dentro del desarrollo moderno de software, DevOps e ingeniería de datos. Su principal objetivo es proporcionar un entorno aislado y reproducible para ejecutar aplicaciones de manera consistente en cualquier infraestructura.

Comprender cómo funcionan los contenedores es fundamental antes de comenzar a trabajar con herramientas como Docker, Kubernetes o plataformas de automatización CI/CD.

## ¿Qué es un Contenedor?

Un contenedor es una unidad estandarizada de software que empaqueta una aplicación junto con todos los recursos necesarios para su ejecución.

Un contenedor incluye:

- Código de la aplicación.
- Librerías requeridas.
- Dependencias.
- Variables de entorno.
- Configuraciones.
- Archivos necesarios para su funcionamiento.

Gracias a este empaquetado, la aplicación puede ejecutarse de forma consistente independientemente del sistema donde se despliegue.

Un contenedor puede visualizarse como una "caja aislada" que contiene todo lo necesario para ejecutar una aplicación.

```text
+----------------------------------+
|           CONTENEDOR             |
+----------------------------------+
| Aplicación                       |
| Dependencias                     |
| Librerías                        |
| Configuraciones                  |
| Variables de entorno             |
+----------------------------------+
```

## Características de los Contenedores

Los contenedores poseen varias características que los han convertido en una tecnología ampliamente adoptada.

### Portabilidad

Un contenedor puede ejecutarse en diferentes entornos sin necesidad de modificar su configuración.

### Aislamiento

Cada contenedor funciona de forma independiente respecto a otros contenedores o aplicaciones instaladas en el mismo sistema.

### Ligereza

No requieren un sistema operativo completo para cada instancia, lo que reduce significativamente el consumo de recursos.

### Consistencia

Garantizan que una aplicación se comporte de la misma forma en todos los entornos.

### Escalabilidad

Es posible crear múltiples instancias de un mismo contenedor de manera rápida y eficiente.

## Arquitectura de un Contenedor

Los contenedores funcionan sobre una arquitectura basada en capas donde cada componente cumple una función específica.

De forma simplificada, la arquitectura puede representarse así:

```text
Aplicación
      │
Dependencias
      │
Imagen del Contenedor
      │
Motor de Contenedores
      │
Sistema Operativo Host
      │
Infraestructura Física o Virtual
```

Cada capa aporta elementos necesarios para que la aplicación pueda ejecutarse correctamente.

## Sistema Operativo Host

El sistema operativo host es el sistema operativo principal donde se ejecutan los contenedores.

Ejemplos comunes:

- Linux
- Windows
- macOS

El host proporciona los recursos físicos que utilizarán los contenedores:

- CPU
- Memoria RAM
- Disco
- Red

A diferencia de las máquinas virtuales, los contenedores comparten el núcleo (Kernel) del sistema operativo anfitrión.

## Kernel Compartido

Uno de los conceptos más importantes de los contenedores es que todos comparten el mismo kernel del sistema operativo host.

Esto significa que no es necesario instalar un sistema operativo completo para cada contenedor.

```text
Sistema Operativo Host
│
├── Kernel Compartido
│
├── Contenedor A
├── Contenedor B
├── Contenedor C
└── Contenedor D
```

Este enfoque permite:

- Menor consumo de recursos.
- Menor tiempo de inicio.
- Mayor densidad de aplicaciones por servidor.

## Espacios de Nombres (Namespaces)

Los Namespaces son mecanismos del sistema operativo Linux que permiten aislar recursos entre contenedores.

Gracias a ellos, cada contenedor tiene su propia visión del sistema.

Los Namespaces aíslan elementos como:

- Procesos.
- Usuarios.
- Interfaces de red.
- Sistemas de archivos.
- Identificadores de procesos.

Por ejemplo, un proceso dentro de un contenedor puede creer que es el proceso principal del sistema, aunque en realidad existan cientos de procesos ejecutándose en el host.

## Control Groups (cgroups)

Los Control Groups o cgroups permiten limitar y administrar los recursos consumidos por cada contenedor.

Entre los recursos que pueden controlarse se encuentran:

- CPU.
- Memoria.
- Disco.
- Uso de red.

Por ejemplo, un contenedor puede configurarse para utilizar únicamente:

- 2 GB de RAM.
- 1 núcleo de CPU.

Esto evita que una aplicación consuma todos los recursos del servidor.

## Sistema de Archivos en Capas

Los contenedores utilizan sistemas de archivos basados en capas.

Cada nueva modificación genera una capa adicional sobre las anteriores.

Por ejemplo:

```text
Capa 4 -> Configuración personalizada
Capa 3 -> Librerías instaladas
Capa 2 -> Python
Capa 1 -> Sistema base Linux
```

Este enfoque ofrece varias ventajas:

- Menor uso de almacenamiento.
- Reutilización de componentes.
- Distribución más eficiente.
- Construcción más rápida.

## Ciclo de Vida de un Contenedor

Un contenedor pasa por diferentes estados durante su existencia.

### Creación

Se genera una nueva instancia basada en una imagen.

### Ejecución

El contenedor inicia sus procesos y comienza a funcionar.

### Pausa

La ejecución puede suspenderse temporalmente.

### Detención

Los procesos internos finalizan.

### Eliminación

El contenedor es removido del sistema.

El ciclo puede representarse de la siguiente manera:

```text
Crear
  │
  ▼
Ejecutar
  │
  ▼
Pausar
  │
  ▼
Reanudar
  │
  ▼
Detener
  │
  ▼
Eliminar
```

Los contenedores pueden utilizarse para múltiples escenarios tecnológicos.

### Aplicaciones Web

### Bases de Datos

### Ingeniería de Datos

### Automatización

- Procesos programados.

## Ventajas frente a Métodos Tradicionales

Comparados con instalaciones tradicionales, los contenedores ofrecen importantes beneficios.

| Característica | Instalación Tradicional | Contenedores |
|---------------|-------------------------|--------------|
| Portabilidad | Baja | Alta |
| Consistencia | Variable | Alta |
| Tiempo de despliegue | Alto | Bajo |
| Escalabilidad | Limitada | Alta |
| Consumo de recursos | Alto | Bajo |
| Facilidad de replicación | Baja | Alta |

## Consideraciones Importantes

Aunque los contenedores ofrecen múltiples ventajas, es importante recordar que son procesos aislados y no máquinas virtuales completas.

Esto implica que:

- Comparten el sistema operativo anfitrión.
- Son más ligeros que una máquina virtual.
- Pueden iniciarse en segundos.
- Son ideales para arquitecturas modernas basadas en microservicios y automatización.

Comprender estos conceptos fundamentales permitirá entender con mayor facilidad el funcionamiento de Docker, las imágenes, la gestión de contenedores y los procesos de automatización que se desarrollarán en los siguientes temas.

# Introducción a Docker



Docker es la plataforma de contenerización más utilizada en la actualidad. Su popularidad se debe a que simplifica la creación, distribución y ejecución de contenedores, permitiendo a los desarrolladores y equipos de operaciones trabajar con entornos consistentes y reproducibles.

Aunque existen otras tecnologías de contenerización, Docker se ha convertido en el estándar de facto para el desarrollo moderno de aplicaciones, la automatización de procesos y la implementación de soluciones en la nube.

## ¿Qué es Docker?

Docker es una plataforma de software que permite crear, administrar y ejecutar contenedores de manera sencilla.

Su objetivo principal es facilitar el empaquetado de aplicaciones junto con todas sus dependencias para que puedan ejecutarse de forma consistente en cualquier entorno compatible.

Docker proporciona herramientas para:

- Crear imágenes de contenedores.
- Ejecutar contenedores.
- Administrar recursos.
- Compartir aplicaciones.
- Automatizar despliegues.

Gracias a Docker, los desarrolladores pueden construir una aplicación una sola vez y ejecutarla en diferentes entornos sin realizar modificaciones.

## Historia de Docker

Docker fue lanzado en 2013 por la empresa Docker Inc. como una solución para simplificar el uso de tecnologías de contenedores basadas en Linux.

Antes de Docker, los contenedores existían mediante tecnologías como:

- LXC (Linux Containers).
- OpenVZ.
- FreeBSD Jails.

Sin embargo, estas herramientas requerían conocimientos avanzados de administración de sistemas.

Docker simplificó enormemente el proceso mediante:

- Interfaces más amigables.
- Automatización de configuraciones.
- Repositorios de imágenes.
- Herramientas de distribución.

Esta simplificación impulsó su adopción masiva en la industria tecnológica.

## Arquitectura de Docker

Docker utiliza una arquitectura cliente-servidor donde diferentes componentes trabajan conjuntamente para administrar los contenedores.

La arquitectura general puede representarse de la siguiente manera:

```text
Usuario
   │
Docker CLI
   │
Docker Engine
   │
├── Imágenes
├── Contenedores
├── Redes
└── Volúmenes
```

Cada componente cumple una función específica dentro del ecosistema Docker.

## Docker Engine

Docker Engine es el núcleo de la plataforma Docker.

Es el componente responsable de:

- Crear contenedores.
- Ejecutar contenedores.
- Administrar imágenes.
- Gestionar redes.
- Controlar almacenamiento.

Cuando un usuario ejecuta un comando Docker, este es procesado por Docker Engine.

Docker Engine está compuesto por tres elementos principales:

### Docker Daemon

Es el servicio que permanece ejecutándose en segundo plano y realiza las operaciones solicitadas.

Responsabilidades:

- Construcción de imágenes.
- Gestión de contenedores.
- Administración de recursos.
- Manejo de redes y volúmenes.

### Docker API

Proporciona una interfaz mediante la cual otras aplicaciones pueden comunicarse con Docker Engine.

Permite:

- Automatización.
- Integración con herramientas externas.
- Desarrollo de soluciones personalizadas.

### Docker CLI

La Command Line Interface (CLI) es la herramienta utilizada por los usuarios para interactuar con Docker.

Ejemplo:

```bash
docker version
```

Este comando solicita información sobre la versión instalada de Docker.

## Docker Desktop

Docker Desktop es una aplicación diseñada para facilitar el uso de Docker en sistemas operativos de escritorio.

Proporciona una interfaz gráfica que simplifica tareas como:

- Gestión de contenedores.
- Visualización de imágenes.
- Monitoreo de recursos.
- Administración de configuraciones.

Docker Desktop está disponible para:

- Windows
- macOS
- Linux

Además, incluye Docker Engine preconfigurado para facilitar la instalación.

## Docker Hub

Docker Hub es el repositorio oficial de imágenes Docker.

Funciona de manera similar a GitHub, pero para imágenes de contenedores.

Permite:

- Descargar imágenes existentes.
- Compartir imágenes propias.
- Gestionar versiones.
- Distribuir aplicaciones.

Algunas imágenes populares disponibles en Docker Hub incluyen:

- Python
- PostgreSQL
- MySQL
- Redis
- Nginx
- Apache Airflow

La mayoría de los proyectos modernos utilizan Docker Hub como punto central para distribuir sus imágenes.

## Flujo de Trabajo de Docker

El proceso típico de trabajo con Docker sigue una serie de pasos bien definidos.

```text
Desarrollador
      │
      ▼
Crear Dockerfile
      │
      ▼
Construir Imagen
      │
      ▼
Ejecutar Contenedor
      │
      ▼
Validar Aplicación
      │
      ▼
Publicar Imagen
      │
      ▼
Desplegar en Producción
```

Este flujo permite garantizar que la aplicación se ejecute de manera consistente durante todo su ciclo de vida.

# Instalación y Configuración de Docker



Antes de comenzar a trabajar con imágenes y contenedores, es necesario instalar y configurar Docker correctamente. Una instalación adecuada garantiza que las herramientas y servicios necesarios estén disponibles para crear, ejecutar y administrar contenedores de forma eficiente.

En este tema se abordará la instalación de Docker en los sistemas operativos más utilizados y las verificaciones básicas necesarias para asegurar que el entorno está listo para trabajar.

## Requisitos de Instalación

Antes de instalar Docker es importante verificar que el equipo cumple con los requisitos mínimos.

### Requisitos Generales

- Procesador de 64 bits.
- Al menos 4 GB de memoria RAM.
- Espacio disponible en disco para imágenes y contenedores.
- Permisos de administrador o superusuario.
- Acceso a Internet para descargar componentes.

### Requisitos para Windows

- Windows 10 o Windows 11.
- Virtualización habilitada en BIOS.
- WSL2 (Windows Subsystem for Linux).

### Requisitos para Linux

- Distribución Linux compatible.
- Kernel actualizado.
- Permisos de usuario sudo.

### Requisitos para macOS

- macOS compatible con Docker Desktop.
- Procesador Intel o Apple Silicon.

## Instalación de Docker en Windows

Actualmente la forma más sencilla de instalar Docker en Windows es mediante Docker Desktop.

### Paso 1: Instalar WSL2

Abrir PowerShell como administrador y ejecutar:

```powershell
wsl --install
```

Reiniciar el equipo al finalizar la instalación.

### Paso 2: Verificar WSL

Ejecutar:

```powershell
wsl --status
```

La salida debe indicar que WSL está correctamente instalado.

### Paso 3: Descargar Docker Desktop

Descargar Docker Desktop desde el sitio oficial de Docker.

### Paso 4: Ejecutar el Instalador

Durante la instalación asegurarse de habilitar:

- Use WSL 2 instead of Hyper-V.
- Add Docker to PATH.

### Paso 5: Reiniciar el Equipo

Una vez finalizada la instalación, reiniciar el sistema.

### Paso 6: Abrir Docker Desktop

Al iniciar Docker Desktop se mostrará el estado del servicio Docker.

Cuando el icono indique que Docker está ejecutándose correctamente, la instalación habrá finalizado.

## Instalación de Docker en Ubuntu

Docker puede instalarse directamente desde los repositorios oficiales.

### Actualizar Paquetes

```bash
sudo apt update
sudo apt upgrade -y
```

### Instalar Dependencias

```bash
sudo apt install -y \
ca-certificates \
curl \
gnupg \
lsb-release
```

### Agregar la Clave Oficial de Docker

```bash
sudo mkdir -p /etc/apt/keyrings

curl -fsSL https://download.docker.com/linux/ubuntu/gpg | \
sudo gpg --dearmor -o /etc/apt/keyrings/docker.gpg
```

### Configurar el Repositorio

```bash
echo \
"deb [arch=$(dpkg --print-architecture) \
signed-by=/etc/apt/keyrings/docker.gpg] \
https://download.docker.com/linux/ubuntu \
$(lsb_release -cs) stable" | \
sudo tee /etc/apt/sources.list.d/docker.list > /dev/null
```

### Instalar Docker

```bash
sudo apt update

sudo apt install -y \
docker-ce \
docker-ce-cli \
containerd.io \
docker-buildx-plugin \
docker-compose-plugin
```

### Verificar el Servicio

```bash
sudo systemctl status docker
```

La salida debe mostrar el servicio en estado activo.

## Verificación de la Instalación

### Verificar la Versión

```bash
docker --version
```

Ejemplo de salida:

# Imágenes Docker



Las imágenes son uno de los componentes más importantes dentro del ecosistema Docker. Todo contenedor se crea a partir de una imagen, por lo que comprender su funcionamiento es fundamental para trabajar correctamente con aplicaciones contenerizadas.

Una imagen puede considerarse como una plantilla o modelo que contiene todo lo necesario para ejecutar una aplicación. A partir de una misma imagen pueden crearse múltiples contenedores independientes.

## ¿Qué es una Imagen Docker?

Una imagen Docker es un paquete inmutable que contiene todos los archivos, dependencias, configuraciones y componentes necesarios para ejecutar una aplicación.

Una imagen puede incluir:

- Sistema operativo base.
- Lenguaje de programación.
- Bibliotecas necesarias.
- Dependencias de la aplicación.
- Archivos de configuración.
- Código fuente.

Cuando se ejecuta una imagen, Docker crea una instancia denominada contenedor.

La relación entre imágenes y contenedores puede compararse con la relación entre una clase y un objeto en programación orientada a objetos.

```text
Imagen Docker
      │
      ├── Contenedor 1
      ├── Contenedor 2
      ├── Contenedor 3
      └── Contenedor 4
```

Una misma imagen puede utilizarse para crear múltiples contenedores.

## Características de las Imágenes

Las imágenes Docker poseen varias características importantes.

### Son Inmutables

Una vez creada una imagen, su contenido no puede modificarse directamente.

Si se requiere un cambio, debe generarse una nueva versión de la imagen.

### Son Portables

Pueden compartirse entre distintos equipos y servidores manteniendo exactamente la misma configuración.

### Son Reutilizables

Una misma imagen puede utilizarse para múltiples proyectos o despliegues.

### Son Versionables

Permiten administrar diferentes versiones de una aplicación mediante etiquetas o tags.

## Arquitectura Basada en Capas

Las imágenes Docker utilizan un sistema de almacenamiento por capas.

Cada instrucción ejecutada durante la construcción de una imagen genera una nueva capa.

Por ejemplo:

```text
Capa 5 → Aplicación
Capa 4 → Librerías Python
Capa 3 → Dependencias del sistema
Capa 2 → Python
Capa 1 → Ubuntu Base
```

Este diseño ofrece importantes ventajas:

- Optimización del almacenamiento.
- Reutilización de componentes.
- Descargas más rápidas.
- Construcciones más eficientes.

Cuando dos imágenes comparten capas, Docker almacena únicamente una copia física de dichas capas.

## Imágenes Base

Toda imagen suele construirse a partir de una imagen base.

Algunas de las más utilizadas son:

| Imagen | Uso |

| node | Aplicaciones Node.js |
| nginx | Servidor web |

```dockerfile
FROM python:3.12
```

La instrucción anterior indica que la imagen utilizará Python 3.12 como base.

# Gestión de Contenedores



Una vez que se dispone de una imagen Docker, el siguiente paso consiste en crear y administrar contenedores. Los contenedores representan instancias en ejecución de una imagen y constituyen el entorno donde realmente se ejecutan las aplicaciones.

La gestión adecuada de contenedores es una de las tareas más frecuentes al trabajar con Docker, ya que permite controlar el ciclo de vida completo de las aplicaciones contenerizadas.

## Relación entre Imagen y Contenedor

Una imagen puede compararse con una plantilla, mientras que un contenedor representa una instancia activa creada a partir de dicha plantilla.

```text
Imagen Python
      │
      ├── Contenedor A
      ├── Contenedor B
      └── Contenedor C
```

Cada contenedor funciona de manera independiente, aunque todos hayan sido creados a partir de la misma imagen.

Por ejemplo, una única imagen de PostgreSQL puede utilizarse para crear múltiples bases de datos aisladas entre sí.

## Creación de Contenedores

La creación de un contenedor se realiza mediante el comando `docker run`.

Sintaxis general:

```bash
docker run [opciones] imagen
```

Ejemplo:

```bash
docker run hello-world
```

Docker realizará automáticamente las siguientes acciones:

1. Buscar la imagen localmente.
2. Descargarla si no existe.
3. Crear el contenedor.
4. Ejecutarlo.
5. Mostrar la salida correspondiente.

# Persistencia de Datos

Uno de los conceptos más importantes al trabajar con contenedores es comprender cómo se almacenan los datos. Por defecto, los contenedores están diseñados para ser temporales y reemplazables, lo que significa que cualquier información almacenada dentro de ellos puede perderse si el contenedor es eliminado.

Para resolver este problema, Docker proporciona mecanismos de persistencia que permiten almacenar información fuera del ciclo de vida del contenedor.

La persistencia de datos es especialmente importante en aplicaciones que generan información continuamente, como bases de datos, sistemas de archivos compartidos, plataformas analíticas y aplicaciones empresariales.

## El Problema del Almacenamiento Efímero

Cuando un contenedor se crea, Docker genera una capa de escritura donde la aplicación puede almacenar archivos y datos.

Por ejemplo:

```text
Contenedor
│
├── Aplicación
├── Configuración
└── Datos Generados
```

Mientras el contenedor exista, estos datos estarán disponibles.

Sin embargo, si el contenedor es eliminado:

```bash
docker rm mi_contenedor
```

Toda la información almacenada dentro de la capa de escritura desaparece.

Este comportamiento es adecuado para aplicaciones temporales, pero representa un problema para sistemas que requieren conservar información.

Por ejemplo:

- Bases de datos.
- Archivos de usuarios.
- Reportes generados.
- Registros históricos.
- Configuraciones personalizadas.

## ¿Qué es la Persistencia de Datos?

La persistencia consiste en almacenar información fuera del contenedor para garantizar que los datos continúen existiendo incluso si el contenedor se detiene, reinicia o elimina.

La arquitectura general puede representarse así:

```text
Contenedor
      │
      ▼
Volumen Docker
      │
      ▼
Disco del Host
```

De esta manera, los datos sobreviven independientemente del estado del contenedor.

## Métodos de Persistencia en Docker

Docker proporciona varios mecanismos para almacenar información de forma permanente.

Los más utilizados son:

- Volúmenes Docker (Volumes).
- Bind Mounts.
- Tmpfs Mounts.

En entornos empresariales los más comunes son los volúmenes y los bind mounts.

## Volúmenes Docker

Los volúmenes son el mecanismo recomendado por Docker para almacenar datos persistentes.

Un volumen es un área de almacenamiento administrada directamente por Docker.

Ventajas:

- Fácil administración.
- Mayor portabilidad.
- Mejor rendimiento.
- Independencia respecto al sistema operativo.

Arquitectura simplificada:

```text
Aplicación
      │
Contenedor
      │
Volumen Docker
      │
Sistema Host
```

## Crear un Volumen

Para crear un volumen:

```bash
docker volume create datos_postgres
```

Ejemplo de salida:

```text
datos_postgres
```

Docker crea automáticamente el espacio de almacenamiento correspondiente.

## Listar Volúmenes

Para visualizar los volúmenes existentes:

```bash
docker volume ls
```

Ejemplo:

```text
DRIVER    VOLUME NAME
local     datos_postgres
local     airflow_data
```

## Inspeccionar un Volumen

Para consultar información detallada:

```bash
docker volume inspect datos_postgres
```

Ejemplo de salida:

```json
[
  {
    "Name": "datos_postgres",
    "Driver": "local",
    "Mountpoint": "/var/lib/docker/volumes/..."
  }
]
```

## Utilizar un Volumen en un Contenedor

Los volúmenes se conectan mediante la opción:

```bash
-v
```

Ejemplo:

```bash
docker run -d \
--name postgres_db \
-v datos_postgres:/var/lib/postgresql/data \
postgres
```

En este caso:

| Elemento | Descripción |
|-----------|-------------|
| datos_postgres | Volumen Docker |
| /var/lib/postgresql/data | Ruta utilizada por PostgreSQL |

Todos los datos de la base de datos serán almacenados en el volumen.

## Verificación de Persistencia

Supongamos que se elimina el contenedor:

```bash
docker rm -f postgres_db
```

Posteriormente se crea uno nuevo utilizando el mismo volumen:

```bash
docker run -d \
--name postgres_db_nuevo \
-v datos_postgres:/var/lib/postgresql/data \
postgres
```

La información continuará disponible porque se encuentra almacenada en el volumen.

## Bind Mounts

Los Bind Mounts permiten conectar directorios específicos del sistema anfitrión con directorios internos del contenedor.

Arquitectura:

```text
Directorio Host
      │
      ▼
Contenedor
```

A diferencia de los volúmenes, el usuario controla directamente la ubicación física de los datos.

## Crear un Bind Mount

Ejemplo:

```bash
docker run -d \
-v /home/usuario/datos:/app/datos \
ubuntu
```

En este caso:

| Host | Contenedor |
|---------|---------|
| /home/usuario/datos | /app/datos |

Los archivos creados en cualquiera de los dos lados estarán disponibles en ambos.

## Caso Práctico con Archivos

Crear una carpeta local:

```bash
mkdir proyecto
```

## Diferencias entre Volúmenes y Bind Mounts

| Característica | Volúmenes | Bind Mounts |
|----------------|-----------|------------|
| Administrado por Docker | Sí | No |
| Portabilidad | Alta | Media |
| Rendimiento | Alto | Variable |
| Dependencia del Host | Baja | Alta |
| Facilidad de respaldo | Alta | Media |

En producción generalmente se prefieren los volúmenes Docker.

Los bind mounts suelen utilizarse durante el desarrollo.

## Tmpfs Mounts

Los Tmpfs Mounts almacenan información únicamente en memoria RAM.

Ejemplo:

```bash
docker run \
--tmpfs /cache \
ubuntu
```

Características:

- No utilizan disco.
- Mayor velocidad.
- Los datos desaparecen al detener el contenedor.

Se utilizan para:

- Cachés temporales.
- Información sensible.
- Procesamiento de alta velocidad.

## Gestión de Volúmenes

### Eliminar un Volumen

```bash
docker volume rm datos_postgres
```

### Eliminar Volúmenes No Utilizados

```bash
docker volume prune
```

Este comando elimina todos los volúmenes que no están asociados a contenedores activos.

## Respaldo de Volúmenes

Es posible realizar respaldos utilizando contenedores temporales.

Ejemplo:

```bash
docker run --rm \
-v datos_postgres:/origen \
-v $(pwd):/backup \
ubuntu \
tar czf /backup/respaldo.tar.gz /origen
```

Este procedimiento genera una copia comprimida del contenido del volumen.

## Restauración de Volúmenes

Para restaurar un respaldo:

```bash
docker run --rm \
-v datos_postgres:/destino \
-v $(pwd):/backup \
ubuntu \
tar xzf /backup/respaldo.tar.gz -C /
```

Este proceso permite recuperar la información almacenada.

## Persistencia en Bases de Datos

Uno de los usos más comunes de los volúmenes es el almacenamiento de bases de datos.

Ejemplo con PostgreSQL:

```bash
docker run -d \
--name postgres_db \
-e POSTGRES_PASSWORD=admin123 \
-v postgres_data:/var/lib/postgresql/data \
postgres
```

Ejemplo con MySQL:

```bash
docker run -d \
--name mysql_db \
-e MYSQL_ROOT_PASSWORD=admin123 \
-v mysql_data:/var/lib/mysql \
mysql
```

Sin volúmenes, toda la información se perdería al eliminar el contenedor.

### Utilizar Volúmenes para Producción

Ofrecen mayor portabilidad y facilidad de administración.

### Mantener Respaldos

Los datos persistentes deben incluirse en la estrategia de respaldo de la organización.

### Nombrar Volúmenes Claramente

Ejemplos:

```text
postgres_data
airflow_logs
jupyter_notebooks
```

### Separar Datos por Aplicación

Evita conflictos y simplifica la administración.

### Limpiar Volúmenes No Utilizados

## Consideraciones Finales

La persistencia de datos es un componente esencial en cualquier arquitectura basada en contenedores. Aunque los contenedores están diseñados para ser temporales, los mecanismos de almacenamiento proporcionados por Docker permiten conservar información de manera segura y eficiente.

El uso adecuado de volúmenes y bind mounts garantiza que las aplicaciones puedan reiniciarse, actualizarse o reemplazarse sin comprometer la integridad de los datos almacenados.

# Redes en Docker

Hasta este punto se ha aprendido a crear imágenes, ejecutar contenedores y almacenar datos de forma persistente. Sin embargo, las aplicaciones modernas rara vez funcionan de manera aislada. Normalmente requieren comunicarse con otros servicios, bases de datos, APIs o aplicaciones distribuidas.

Docker proporciona un sistema de redes que permite conectar contenedores entre sí y facilitar la comunicación con sistemas externos.

Comprender el funcionamiento de las redes es fundamental para construir arquitecturas multicontenedor, microservicios y plataformas de datos modernas.

## ¿Por Qué Son Necesarias las Redes?

Supongamos que se tiene una aplicación web y una base de datos ejecutándose en contenedores diferentes.

```text
Aplicación Web
       │
       ▼
Base de Datos
```

Para que la aplicación pueda consultar información, ambos contenedores deben poder comunicarse.

Las redes Docker permiten:

- Comunicación entre contenedores.
- Acceso desde el host hacia los contenedores.
- Comunicación con sistemas externos.
- Aislamiento de servicios.
- Organización de aplicaciones complejas.

## Conceptos Básicos de Networking

Cuando un contenedor se crea, Docker le asigna:

- Una dirección IP.
- Una interfaz de red virtual.
- Un nombre interno.
- Reglas de comunicación.

Por ejemplo:

```text
Contenedor Web
IP: 172.18.0.2

Contenedor PostgreSQL
IP: 172.18.0.3
```

Estos contenedores pueden intercambiar información utilizando la red configurada.

## Redes Predeterminadas en Docker

Docker crea automáticamente varias redes durante la instalación.

Para visualizarlas:

```bash
docker network ls
```

Ejemplo:

```text
NETWORK ID     NAME      DRIVER
abc123         bridge    bridge
def456         host      host
ghi789         none      null
```

Las redes más importantes son:

- bridge
- host
- none

## Red Bridge

La red bridge es la red predeterminada utilizada por Docker.

Cuando se ejecuta un contenedor sin especificar una red, Docker lo conecta automáticamente a esta red.

Ejemplo:

```bash
docker run -d nginx
```

Docker asignará automáticamente una dirección IP dentro de la red bridge.

Arquitectura simplificada:

```text
Docker Host
│
├── Bridge Network
│
├── Contenedor A
├── Contenedor B
└── Contenedor C
```

Los contenedores pueden comunicarse utilizando sus direcciones IP internas.

## Inspeccionar la Red Bridge

Para obtener información detallada:

```bash
docker network inspect bridge
```

La salida incluye:

- Configuración IP.
- Contenedores conectados.
- Subred asignada.
- Gateway.

Ejemplo:

```json
{
  "Subnet": "172.17.0.0/16",
  "Gateway": "172.17.0.1"
}
```

## Publicación de Puertos

Por defecto, los servicios dentro de un contenedor no son accesibles desde el exterior.

Para exponer un servicio se utiliza:

```bash
-p
```

Sintaxis:

```bash
docker run -p PUERTO_HOST:PUERTO_CONTENEDOR
```

Ejemplo:

```bash
docker run -d -p 8080:80 nginx
```

Resultado:

```text
Host: 8080
Contenedor: 80
```

Ahora el servidor web podrá accederse mediante:

```text
http://localhost:8080
```

## Ejemplo con PostgreSQL

```bash
docker run -d \
--name postgres_db \
-e POSTGRES_PASSWORD=admin123 \
-p 5432:5432 \
postgres
```

Con esta configuración cualquier herramienta cliente podrá conectarse utilizando:

```text
localhost:5432
```

## Red Host

La red host permite que el contenedor utilice directamente la red del sistema anfitrión.

Ejemplo:

```bash
docker run --network host nginx
```

Arquitectura:

```text
Sistema Host
      │
Contenedor
```

Ventajas:

- Mayor rendimiento.
- Menor sobrecarga de red.

Desventajas:

- Menor aislamiento.
- Posibles conflictos de puertos.

Generalmente se utiliza en entornos Linux.

## Red None

La red none deshabilita completamente la conectividad.

Ejemplo:

```bash
docker run --network none ubuntu
```

Resultado:

```text
Sin acceso a red
```

Casos de uso:

- Procesamiento aislado.
- Seguridad.
- Pruebas controladas.

## Redes Personalizadas

Por ejemplo:

```text
Contenedor Web
      │
      ▼
postgres_db
```

La aplicación puede conectarse utilizando:

```text
postgres_db
```

sin necesidad de conocer la dirección IP.

Esta característica simplifica enormemente la configuración de aplicaciones multicontenedor.

Docker agregará el contenedor a la red indicada.

## Desconectar un Contenedor

El contenedor dejará de participar en esa red.

## Eliminación de Redes

Para eliminar una red:

```bash
docker network rm red_aplicacion
```

La red debe encontrarse sin contenedores asociados.

## Redes Overlay

Las redes overlay permiten la comunicación entre contenedores distribuidos en múltiples servidores.

Arquitectura simplificada:

```text
Servidor 1
│
├── Contenedor A
│
└── Overlay Network
          │
Servidor 2
│
├── Contenedor B
│
└── Overlay Network
```

Estas redes suelen utilizarse en plataformas de orquestación avanzadas.

## Redes Macvlan

Macvlan permite asignar direcciones IP físicas directamente a los contenedores.

Ejemplo conceptual:

```text
Contenedor A → 192.168.1.50
Contenedor B → 192.168.1.51
```

Cada contenedor aparece como un dispositivo independiente dentro de la red corporativa.

Este tipo de red suele utilizarse en escenarios especializados.

Facilitan la administración y mejoran la seguridad.

Las redes Docker proporcionan la infraestructura necesaria para que múltiples contenedores trabajen conjuntamente como una solución integrada. Gracias a ellas es posible construir aplicaciones distribuidas, plataformas de datos y arquitecturas multicontenedor de manera sencilla y eficiente.

# Dockerfile

Hasta este momento se han utilizado imágenes descargadas desde Docker Hub. Sin embargo, en proyectos reales es necesario crear imágenes personalizadas que incluyan aplicaciones propias, configuraciones específicas y dependencias particulares.

Docker proporciona un mecanismo denominado **Dockerfile**, que permite definir de forma automatizada cómo construir una imagen.

Un Dockerfile es el punto de partida para empaquetar aplicaciones dentro de contenedores y constituye una de las herramientas más importantes dentro del ecosistema Docker.

## ¿Qué es un Dockerfile?

Un Dockerfile es un archivo de texto que contiene una serie de instrucciones utilizadas por Docker para construir una imagen.

Estas instrucciones describen:

- Imagen base.
- Dependencias necesarias.
- Archivos que deben copiarse.
- Variables de entorno.
- Comandos de ejecución.
- Configuración del contenedor.

Por ejemplo:

```dockerfile
FROM python:3.12

WORKDIR /app

COPY . .

RUN pip install -r requirements.txt

CMD ["python", "app.py"]
```

A partir de este archivo Docker puede generar automáticamente una imagen lista para ejecutarse.

## Beneficios del Dockerfile

Utilizar Dockerfiles ofrece múltiples ventajas.

### Automatización

Permite construir imágenes de manera repetible y consistente.

### Versionamiento

Puede almacenarse junto con el código fuente en Git.

### Portabilidad

Garantiza que cualquier desarrollador pueda generar exactamente la misma imagen.

### Reproducibilidad

Evita configuraciones manuales y errores humanos.

### Integración con CI/CD

Permite automatizar completamente la construcción de imágenes.

## Estructura General de un Dockerfile

Un Dockerfile suele seguir una estructura similar a la siguiente:

```dockerfile
FROM imagen_base

WORKDIR directorio

COPY archivos

RUN comandos

EXPOSE puerto

CMD comando_inicio
```

Cada instrucción cumple una función específica dentro del proceso de construcción.

## Instrucción FROM

La instrucción FROM define la imagen base sobre la cual se construirá la nueva imagen.

Ejemplo:

```dockerfile
FROM python:3.12
```

También puede utilizarse:

```dockerfile
FROM ubuntu:24.04
```

o

```dockerfile
FROM alpine:latest
```

Todo Dockerfile debe comenzar con una instrucción FROM.

## Instrucción WORKDIR

WORKDIR establece el directorio de trabajo dentro del contenedor.

Ejemplo:

```dockerfile
WORKDIR /app
```

A partir de ese momento todas las operaciones se ejecutarán dentro de esa carpeta.

Equivale conceptualmente a:

```bash
cd /app
```

## Instrucción COPY

COPY permite copiar archivos desde el equipo local hacia la imagen.

Ejemplo:

```dockerfile
COPY . .
```

Significado:

| Origen | Destino |
|----------|----------|
| Directorio actual | Directorio de trabajo |

También puede utilizarse:

```dockerfile
COPY app.py /app
```

o

```dockerfile
COPY requirements.txt /app
```

## Instrucción ADD

ADD es similar a COPY pero ofrece funcionalidades adicionales.

Ejemplo:

```dockerfile
ADD proyecto.zip /app
```

También puede descargar archivos remotos.

Sin embargo, actualmente se recomienda utilizar COPY cuando sea posible debido a su comportamiento más predecible.

## Instrucción RUN

RUN ejecuta comandos durante la construcción de la imagen.

Ejemplo:

```dockerfile
RUN pip install pandas
```

Otro ejemplo:

```dockerfile
RUN apt-get update
```

También pueden encadenarse múltiples comandos:

```dockerfile
RUN apt-get update && \
    apt-get install -y curl
```

Cada instrucción RUN genera una nueva capa dentro de la imagen.

## Instrucción ENV

ENV permite definir variables de entorno.

Ejemplo:

```dockerfile
ENV APP_ENV=production
```

Otra variable:

```dockerfile
ENV PORT=8000
```

Posteriormente la aplicación podrá acceder a estos valores.

Ejemplo en Python:

```python
import os

entorno = os.getenv("APP_ENV")
print(entorno)
```

## Instrucción EXPOSE

EXPOSE documenta los puertos utilizados por la aplicación.

Ejemplo:

```dockerfile
EXPOSE 8000
```

o

```dockerfile
EXPOSE 5432
```

Esta instrucción no publica puertos automáticamente.

Su propósito principal es documentar la configuración de la imagen.

## Instrucción CMD

CMD define el comando que se ejecutará al iniciar el contenedor.

Ejemplo:

```dockerfile
CMD ["python", "app.py"]
```

Otro ejemplo:

```dockerfile
CMD ["jupyter", "notebook"]
```

Solo puede existir una instrucción CMD efectiva por imagen.

Si se especifican varias, Docker utilizará la última.

## Instrucción ENTRYPOINT

ENTRYPOINT también define el proceso principal del contenedor.

Ejemplo:

```dockerfile
ENTRYPOINT ["python"]
```

Posteriormente:

```bash
docker run mi_imagen script.py
```

Docker ejecutará:

```bash
python script.py
```

ENTRYPOINT suele utilizarse cuando se desea que un comando específico siempre sea ejecutado.

## Primer Dockerfile para Python

Supongamos la siguiente aplicación.

Archivo:

```python
print("Hola desde Docker")
```

Nombre:

```text
app.py
```

Dockerfile:

```dockerfile
FROM python:3.12

WORKDIR /app

COPY app.py .

CMD ["python", "app.py"]
```

## Construcción de una Imagen

Para construir una imagen:

```bash
docker build -t mi_python_app .
```

Parámetros:

| Elemento | Descripción |
|-----------|-------------|
| build | Construye la imagen |
| -t | Asigna nombre |
| . | Contexto actual |

Resultado:

```text
Successfully built
```

## Verificación de la Imagen

Consultar imágenes disponibles:

```bash
docker images
```

Ejemplo:

```text
REPOSITORY      TAG
mi_python_app   latest
```

## Ejecución de la Imagen

Crear un contenedor:

```bash
docker run mi_python_app
```

Salida:

```text
Hola desde Docker
```

La aplicación estará completamente contenerizada.

## Dockerfile para Aplicaciones con Dependencias

Supongamos una aplicación que utiliza Pandas.

Archivo:

```text
requirements.txt
```

Contenido:

```text
pandas
numpy
```

Dockerfile:

```dockerfile
FROM python:3.12

WORKDIR /app

COPY requirements.txt .

RUN pip install -r requirements.txt

COPY . .

CMD ["python", "app.py"]
```

Durante la construcción Docker instalará automáticamente todas las dependencias.

## Caché de Construcción

Docker almacena capas en caché para acelerar futuras construcciones.

Ejemplo:

```dockerfile
COPY requirements.txt .
RUN pip install -r requirements.txt

COPY . .
```

Ventaja:

Si únicamente cambia el código fuente, Docker reutilizará la instalación de dependencias existente.

Esto reduce significativamente los tiempos de construcción.

## Archivo .dockerignore

Al igual que Git utiliza `.gitignore`, Docker utiliza `.dockerignore`.

Ejemplo:

```text
__pycache__
.git
.env
venv
*.log
```

Beneficios:

- Reduce tamaño de la imagen.
- Acelera construcciones.
- Evita copiar archivos innecesarios.

## Buenas Prácticas para Dockerfiles

### Utilizar Imágenes Base Específicas

Incorrecto:

```dockerfile
FROM python:latest
```

Correcto:

```dockerfile
FROM python:3.12
```

### Minimizar el Número de Capas

Agrupar instrucciones relacionadas.

Incorrecto:

```dockerfile
RUN apt-get update
RUN apt-get install curl
```

Correcto:

```dockerfile
RUN apt-get update && \
    apt-get install -y curl
```

### Utilizar .dockerignore

Reduce tamaño y mejora rendimiento.

### Instalar Solo Dependencias Necesarias

Evita imágenes excesivamente grandes.

### Mantener Dockerfiles Simples

Facilita mantenimiento y comprensión.

## Caso de Uso en Ingeniería de Datos

Supongamos un proceso ETL desarrollado en Python.

Estructura:

```text
etl/
│
├── etl.py
├── requirements.txt
└── Dockerfile
```

Dockerfile:

```dockerfile
FROM python:3.12

WORKDIR /etl

COPY requirements.txt .

RUN pip install -r requirements.txt

COPY . .

CMD ["python", "etl.py"]
```

Con esta configuración el proceso ETL puede ejecutarse en cualquier servidor compatible con Docker sin necesidad de instalar Python o librerías manualmente.

## Flujo de Trabajo Completo

```text
Código Fuente
       │
       ▼
Dockerfile
       │
       ▼
docker build
       │
       ▼
Imagen
       │
       ▼
docker run
       │
       ▼
Contenedor
```

Este flujo constituye la base de prácticamente todas las aplicaciones contenerizadas modernas.

## Consideraciones Finales

El Dockerfile es el mecanismo que permite transformar una aplicación en una imagen Docker reutilizable y portable. Gracias a él es posible automatizar completamente la construcción de entornos de ejecución consistentes, facilitando el desarrollo, las pruebas y los despliegues.

Dominar las instrucciones fundamentales de un Dockerfile es un paso esencial para trabajar profesionalmente con Docker, ya que todas las imágenes personalizadas utilizadas en proyectos reales se construyen a partir de este archivo.

# Docker Compose

A medida que las aplicaciones crecen en complejidad, es común que requieran múltiples servicios funcionando simultáneamente. Por ejemplo, una aplicación web puede necesitar una base de datos, un sistema de caché y un servicio de procesamiento en segundo plano.

Administrar cada contenedor individualmente mediante comandos Docker puede volverse complicado y difícil de mantener. Para resolver este problema surge Docker Compose, una herramienta que permite definir y ejecutar aplicaciones multicontenedor utilizando un único archivo de configuración.

Docker Compose simplifica la administración de entornos completos y facilita la reproducción de configuraciones en diferentes equipos y servidores.

## Problema de las Aplicaciones Multicontenedor

Supongamos una aplicación compuesta por:

- Aplicación Python.
- PostgreSQL.
- Redis.

Sin Docker Compose sería necesario ejecutar múltiples comandos:

```bash
docker network create proyecto_red

docker run ...

docker run ...

docker run ...
```

Además, habría que administrar:

- Redes.
- Volúmenes.
- Variables de entorno.
- Dependencias entre servicios.

Conforme aumenta el número de componentes, la administración manual se vuelve compleja.

## ¿Qué es Docker Compose?

Docker Compose es una herramienta que permite definir y administrar múltiples contenedores mediante un archivo YAML.

Con Docker Compose es posible:

- Crear contenedores.
- Configurar redes.
- Gestionar volúmenes.
- Definir variables de entorno.
- Controlar dependencias entre servicios.

Todo desde un único archivo denominado:

```text
docker-compose.yml
```

o en versiones modernas:

```text
compose.yml
```

## Beneficios de Docker Compose

### Automatización

Permite iniciar entornos completos con un solo comando.

### Reproducibilidad

Todos los desarrolladores trabajan sobre la misma configuración.

### Simplicidad

Reduce significativamente la cantidad de comandos necesarios.

### Escalabilidad

Facilita la administración de aplicaciones complejas.

### Portabilidad

Permite trasladar entornos completos entre diferentes equipos.

## Estructura de un Archivo Compose

Un archivo Compose está escrito en formato YAML.

Ejemplo básico:

```yaml
services:
  web:
    image: nginx
```

Este archivo define un único servicio basado en la imagen de Nginx.

La estructura general suele incluir:

```yaml
services:
volumes:
networks:
```

## Primer Ejemplo de Docker Compose

Archivo:

```yaml
services:
  web:
    image: nginx
    ports:
      - "8080:80"
```

Guardar como:

```text
compose.yml
```

Iniciar servicios:

```bash
docker compose up
```

Resultado:

```text
Nginx disponible en localhost:8080
```

## Comando docker compose up

Este comando crea y ejecuta todos los servicios definidos.

```bash
docker compose up
```

Proceso realizado:

1. Lee el archivo Compose.
2. Descarga imágenes necesarias.
3. Crea redes.
4. Crea volúmenes.
5. Inicia los contenedores.

## Ejecución en Segundo Plano

Para ejecutar los servicios en background:

```bash
docker compose up -d
```

El parámetro `-d` significa detached mode.

Los contenedores continuarán ejecutándose aunque la terminal sea cerrada.

## Detener Servicios

Para detener todos los servicios:

```bash
docker compose down
```

# Contenerización de Aplicaciones Python



Una de las aplicaciones más comunes de Docker es la contenerización de proyectos Python. Gracias a esta práctica es posible distribuir aplicaciones completas junto con todas sus dependencias, garantizando que funcionen de manera consistente en cualquier entorno.

La contenerización elimina problemas relacionados con versiones de Python, bibliotecas incompatibles y configuraciones específicas del sistema operativo, facilitando enormemente el desarrollo, las pruebas y el despliegue de aplicaciones.

## ¿Por Qué Contenerizar Aplicaciones Python?

Tradicionalmente, para ejecutar una aplicación Python era necesario instalar manualmente:

- Python.
- Dependencias.
- Variables de entorno.
- Configuraciones adicionales.

Por ejemplo:

```bash
pip install pandas
pip install numpy
pip install requests
```

En proyectos grandes esto puede generar problemas como:

- Dependencias incompatibles.
- Diferencias entre entornos.
- Dificultades de despliegue.
- Errores de configuración.

Docker resuelve estos problemas encapsulando toda la aplicación dentro de una imagen reproducible.

## Estructura Básica de un Proyecto Python

Supongamos el siguiente proyecto:

```text
proyecto_python/
│
├── app.py
├── requirements.txt
└── Dockerfile
```

Archivo:

```python
print("Hola desde Docker")
```

Guardar como:

```text
app.py
```

## Archivo requirements.txt

Este archivo contiene las dependencias necesarias.

Ejemplo:

```text
pandas
numpy
requests
```

Las bibliotecas especificadas serán instaladas automáticamente durante la construcción de la imagen.

## Dockerfile para una Aplicación Simple

Crear el archivo:

```text
Dockerfile
```

Contenido:

```dockerfile
FROM python:3.12

WORKDIR /app

COPY requirements.txt .

RUN pip install -r requirements.txt

COPY . .

CMD ["python", "app.py"]
```

Este Dockerfile realiza las siguientes acciones:

1. Utiliza Python 3.12 como imagen base.
2. Define el directorio de trabajo.
3. Copia las dependencias.
4. Instala las librerías necesarias.
5. Copia el código fuente.
6. Ejecuta la aplicación.

## Construcción de la Imagen

Desde la raíz del proyecto:

```bash
docker build -t python-demo .
```

Resultado esperado:

```text
Successfully built
Successfully tagged python-demo
```

Verificación:

```bash
docker images
```

Salida:

```text
REPOSITORY    TAG
python-demo   latest
```

Una vez creada la imagen:

Resultado:

```text
Hola desde Docker
```

La aplicación se ejecutará exactamente igual en cualquier equipo que disponga de Docker.

Incorrecto:

La primera opción aprovecha mejor el sistema de caché de Docker.

# Introducción a CI/CD

A medida que los proyectos de software crecen, también aumenta la complejidad de las tareas relacionadas con pruebas, integración, construcción y despliegue. Realizar estos procesos manualmente puede generar errores, retrasos y dificultades para mantener la calidad del software.

Para resolver estos desafíos surgieron las prácticas de **CI/CD (Continuous Integration / Continuous Delivery / Continuous Deployment)**, cuyo objetivo es automatizar gran parte del ciclo de vida del desarrollo de software.

Actualmente, CI/CD es una práctica fundamental en áreas como desarrollo de software, DevOps, ingeniería de datos y plataformas cloud, ya que permite entregar cambios de manera rápida, segura y consistente.

## ¿Qué es CI/CD?

CI/CD es un conjunto de prácticas y metodologías orientadas a automatizar el proceso de desarrollo, validación y despliegue de aplicaciones.

El término está compuesto por tres conceptos principales:

- Continuous Integration (CI).
- Continuous Delivery (CD).
- Continuous Deployment (CD).

Aunque suelen mencionarse conjuntamente, cada uno cumple un propósito diferente dentro del ciclo de desarrollo.

## Problemas del Desarrollo Tradicional

Antes de la adopción de CI/CD, los equipos enfrentaban diversos problemas.

Por ejemplo:

```text
Desarrollador A
     │
Desarrollador B
     │
Desarrollador C
```

Cada integrante trabajaba de forma independiente y los cambios se integraban únicamente al finalizar el desarrollo.

Esto generaba situaciones como:

- Conflictos de código.
- Errores difíciles de detectar.
- Despliegues manuales.
- Procesos lentos.
- Ambientes inconsistentes.

Además, las pruebas normalmente se realizaban al final del proyecto, cuando corregir errores era mucho más costoso.

## Objetivos de CI/CD

Las prácticas CI/CD buscan resolver estos problemas mediante automatización.

Los principales objetivos son:

- Detectar errores rápidamente.
- Reducir tareas manuales.
- Aumentar la calidad del software.
- Acelerar la entrega de nuevas funcionalidades.
- Garantizar despliegues consistentes.
- Facilitar el trabajo colaborativo.

## Concepto de Integración Continua (CI)

La Integración Continua consiste en integrar frecuentemente los cambios realizados por los desarrolladores dentro de un repositorio compartido.

Cada vez que un desarrollador realiza una actualización:

```text
Git Push
    │
    ▼
Repositorio
    │
    ▼
Validación Automática
```

Se ejecutan procesos automáticos que verifican que el código funcione correctamente.

Estas validaciones pueden incluir:

- Revisión de sintaxis.
- Ejecución de pruebas.
- Análisis de calidad.
- Construcción de artefactos.

## Beneficios de la Integración Continua

### Detección Temprana de Errores

Los problemas se identifican poco después de ser introducidos.

### Menor Riesgo de Integración

Las integraciones frecuentes reducen conflictos complejos.

### Retroalimentación Rápida

Los desarrolladores reciben resultados casi inmediatamente.

### Mayor Calidad del Código

Las validaciones automáticas permiten mantener estándares consistentes.

## Flujo de Integración Continua

Un flujo típico puede representarse de la siguiente manera:

```text
Desarrollador
      │
      ▼
Git Push
      │
      ▼
Repositorio
      │
      ▼
Pipeline CI
      │
      ├── Validación
      ├── Pruebas
      └── Construcción
      │
      ▼
Resultado
```

Si alguna etapa falla, el cambio puede corregirse antes de avanzar.

## Concepto de Entrega Continua (Continuous Delivery)

La Entrega Continua amplía la Integración Continua incorporando la preparación automática para despliegues.

Una vez que el código supera las validaciones:

```text
Código Validado
       │
       ▼
Artefacto Generado
       │
       ▼
Listo para Producción
```

El software queda preparado para ser desplegado en cualquier momento.

La decisión final de desplegar continúa siendo responsabilidad de una persona o equipo.

## Beneficios de la Entrega Continua

### Menor Tiempo de Liberación

Las nuevas versiones pueden publicarse rápidamente.

### Mayor Confiabilidad

Los artefactos han pasado por procesos automáticos de validación.

### Menor Riesgo

Las liberaciones son más pequeñas y controladas.

### Automatización del Proceso

Se eliminan múltiples tareas manuales.

## Flujo de Entrega Continua

```text
Código Fuente
      │
      ▼
Pipeline CI
      │
      ▼
Pruebas
      │
      ▼
Build
      │
      ▼
Artefacto
      │
      ▼
Aprobación Manual
      │
      ▼
Producción
```

La aprobación manual diferencia la entrega continua del despliegue continuo.

## Concepto de Despliegue Continuo (Continuous Deployment)

El Despliegue Continuo lleva la automatización un paso más allá.

Una vez que el software supera todas las validaciones:

```text
Pruebas Exitosas
        │
        ▼
Producción
```

La liberación ocurre automáticamente sin intervención humana.

Cada cambio aprobado termina desplegado en producción.

## Beneficios del Despliegue Continuo

### Publicaciones Más Frecuentes

Permite liberar cambios varias veces al día.

### Menor Tiempo de Espera

Los usuarios reciben nuevas funcionalidades más rápidamente.

### Automatización Completa

Reduce significativamente las tareas operativas.

### Retroalimentación Inmediata

Permite validar rápidamente el impacto de los cambios.

## Diferencia entre Delivery y Deployment

Aunque los términos suelen confundirse, existe una diferencia importante.

### Continuous Delivery

```text
Pipeline Completo
       │
       ▼
Aprobación Manual
       │
       ▼
Producción
```

### Continuous Deployment

```text
Pipeline Completo
       │
       ▼
Producción Automática
```

La principal diferencia radica en la existencia o no de una aprobación humana antes del despliegue.

## Pipeline CI/CD

Un pipeline es una secuencia automatizada de pasos que permiten validar y desplegar una aplicación.

Ejemplo:

```text
Código Fuente
      │
      ▼
Compilación
      │
      ▼
Pruebas
      │
      ▼
Análisis
      │
      ▼
Empaquetado
      │
      ▼
Despliegue
```

Cada etapa debe completarse exitosamente para avanzar a la siguiente.

## Ejemplo de Pipeline para Python

Supongamos una API desarrollada en Python.

Pipeline:

```text
Git Push
    │
    ▼
Instalar Dependencias
    │
    ▼
Ejecutar Tests
    │
    ▼
Construir Imagen Docker
    │
    ▼
Publicar Imagen
```

Todo este proceso puede ejecutarse automáticamente sin intervención manual.

## Consideraciones Finales

CI/CD representa una de las prácticas más importantes dentro del desarrollo moderno de software y la ingeniería de datos. Su objetivo principal es automatizar las tareas de integración, validación y despliegue para reducir errores y acelerar la entrega de valor.

# Componentes de un Pipeline CI/CD



Un pipeline CI/CD es una secuencia automatizada de procesos que permite transformar cambios realizados en el código fuente en aplicaciones listas para ser desplegadas. Su función principal es automatizar tareas repetitivas, garantizar la calidad del software y acelerar la entrega de nuevas versiones.

Aunque la implementación puede variar según la herramienta utilizada, la mayoría de los pipelines modernos están compuestos por una serie de componentes fundamentales que trabajan de forma coordinada.

Comprender estos componentes es esencial para diseñar procesos de automatización eficientes y mantenibles.

## ¿Qué es un Pipeline?

Un pipeline puede entenderse como una cadena de etapas que se ejecutan de forma secuencial o paralela.

Cada etapa realiza una tarea específica y únicamente permite avanzar a la siguiente si la ejecución fue exitosa.

Ejemplo conceptual:

```text
Código Fuente
      │
      ▼
Validación
      │
      ▼
Pruebas
      │
      ▼
Construcción
      │
      ▼
Publicación
      │
      ▼
Despliegue
```

Si alguna etapa falla, el pipeline se detiene y reporta el problema.

## Repositorio de Código Fuente

El punto de partida de cualquier pipeline es el repositorio donde se almacena el código.

Algunas plataformas comunes son:

- GitHub
- GitLab
- Bitbucket
- Azure Repos

El repositorio contiene:

- Código fuente.
- Configuraciones.
- Dockerfiles.
- Scripts.
- Archivos de automatización.

Ejemplo:

```text
Proyecto
│
├── src/
├── tests/
├── Dockerfile
└── requirements.txt
```

Cada cambio enviado al repositorio puede activar automáticamente el pipeline.

## Eventos Disparadores (Triggers)

Los triggers son eventos que inician la ejecución del pipeline.

Algunos ejemplos comunes:

### Push

Cuando un desarrollador envía cambios.

```text
git push
```

### Pull Request

Cuando se solicita fusionar cambios entre ramas.

### Creación de Tags

Cuando se genera una nueva versión.

### Ejecución Manual

Cuando un usuario inicia el proceso manualmente.

### Programación

Cuando el pipeline se ejecuta según una fecha u horario específico.

Por ejemplo:

```text
Todos los días a las 2:00 AM
```

## Etapa de Validación

La validación es normalmente la primera fase del pipeline.

Su objetivo es verificar que el código cumple con los requisitos mínimos antes de continuar.

Ejemplos:

- Sintaxis correcta.
- Formato de archivos.
- Estructura del proyecto.
- Archivos obligatorios.

En proyectos Python pueden utilizarse herramientas como:

```bash
python -m py_compile
```

o

```bash
flake8
```

Esta etapa permite detectar errores rápidamente.

## Etapa de Pruebas

Una vez validado el código, el pipeline ejecuta pruebas automatizadas.

Las pruebas permiten verificar que la aplicación continúa funcionando correctamente.

Tipos comunes:

### Pruebas Unitarias

Validan funciones individuales.

### Pruebas de Integración

Validan la interacción entre componentes.

### Pruebas Funcionales

Validan funcionalidades completas.

### Pruebas de Rendimiento

Miden el comportamiento bajo carga.

Ejemplo:

```bash
pytest
```

Si alguna prueba falla, el pipeline se detiene.

## Etapa de Construcción (Build)

La fase de construcción genera los artefactos necesarios para ejecutar o distribuir la aplicación.

Dependiendo del proyecto, los artefactos pueden ser:

- Ejecutables.
- Paquetes.
- Librerías.
- Imágenes Docker.

Ejemplo:

```bash
docker build -t app:v1 .
```

Resultado:

```text
Imagen Docker lista para distribución
```

Esta etapa transforma el código fuente en un producto utilizable.

## Artefactos

Un artefacto es cualquier archivo generado durante el pipeline que será utilizado posteriormente.

Ejemplos:

- Archivos ZIP.
- Ejecutables.
- Paquetes Python.
- Imágenes Docker.
- Reportes.

Representación:

```text
Código Fuente
      │
      ▼
Build
      │
      ▼
Artefacto
```

Los artefactos suelen almacenarse temporalmente para etapas posteriores.

## Repositorios de Artefactos

Una vez generados, los artefactos pueden almacenarse en repositorios especializados.

Ejemplos:

### Docker Hub

Para imágenes Docker.

### GitHub Container Registry

Para contenedores asociados a GitHub.

### Amazon ECR

Repositorio de contenedores de AWS.

### Azure Container Registry

Repositorio de Microsoft Azure.

Ejemplo:

```text
app:v1.0.0
```

Disponible para futuros despliegues.

## Etapa de Análisis de Calidad

Muchos pipelines incluyen verificaciones adicionales relacionadas con la calidad del código.

Ejemplos:

- Complejidad del código.
- Cobertura de pruebas.
- Vulnerabilidades.
- Convenciones de programación.

Herramientas comunes:

```text
SonarQube
Bandit
Pylint
```

Estas verificaciones ayudan a mantener estándares consistentes.

## Gestión de Dependencias

Los pipelines suelen instalar automáticamente las dependencias requeridas por el proyecto.

Ejemplo:

```bash
pip install -r requirements.txt
```

Este proceso garantiza que las pruebas y construcciones se ejecuten en condiciones controladas.

## Gestión de Variables

Los pipelines suelen utilizar variables para parametrizar configuraciones.

Ejemplos:

```text
APP_ENV=production
VERSION=1.0.0
PORT=8000
```

Beneficios:

- Reutilización.
- Flexibilidad.
- Configuración dinámica.

Las variables pueden utilizarse en múltiples etapas del proceso.

## Gestión de Secretos

Algunos procesos requieren credenciales o información sensible.

Ejemplos:

- Contraseñas.
- Tokens.
- Claves API.
- Credenciales cloud.

Ejemplo conceptual:

```text
DATABASE_PASSWORD
AWS_ACCESS_KEY
```

Los secretos nunca deben almacenarse directamente en el código fuente.

Las plataformas CI/CD proporcionan mecanismos seguros para administrarlos.

## Entornos de Despliegue

Los pipelines suelen trabajar con distintos entornos.

### Desarrollo

Utilizado por los desarrolladores.

### Pruebas

Utilizado para validaciones.

### Staging

Replica el entorno de producción.

### Producción

Entorno utilizado por los usuarios finales.

Representación:

```text
Desarrollo
      │
      ▼
Pruebas
      │
      ▼
Staging
      │
      ▼
Producción
```

Cada entorno puede tener configuraciones diferentes.

## Etapa de Despliegue

La fase de despliegue publica la aplicación en el entorno correspondiente.

Ejemplos:

- Servidores físicos.
- Máquinas virtuales.
- Contenedores.
- Plataformas cloud.

Representación:

```text
Artefacto
      │
      ▼
Servidor
```

o

```text
Imagen Docker
      │
      ▼
Cloud Run
```

El objetivo es poner la aplicación a disposición de los usuarios.

## Notificaciones

Muchos pipelines generan notificaciones automáticas.

Ejemplos:

- Correo electrónico.
- Slack.
- Microsoft Teams.
- Discord.

Eventos notificados:

- Éxito.
- Fallo.
- Despliegue completado.

Esto facilita el monitoreo de los procesos.

## Monitoreo y Auditoría

Los pipelines generan registros detallados de cada ejecución.

Ejemplo:

```text
Inicio: 10:15
Pruebas: OK
Build: OK
Deploy: OK
Fin: 10:18
```

Beneficios:

- Trazabilidad.
- Diagnóstico de errores.
- Auditoría de cambios.

## Pipeline Completo

Un pipeline moderno puede representarse de la siguiente forma:

```text
Repositorio
      │
      ▼
Trigger
      │
      ▼
Validación
      │
      ▼
Pruebas
      │
      ▼
Análisis
      │
      ▼
Build
      │
      ▼
Artefacto
      │
      ▼
Repositorio de Artefactos
      │
      ▼
Despliegue
      │
      ▼
Producción
```

Cada etapa aporta controles que incrementan la calidad y confiabilidad del proceso.

## Caso de Uso en Ingeniería de Datos

Supongamos un proyecto ETL desarrollado en Python.

Pipeline:

```text
Git Push
      │
      ▼
Instalar Dependencias
      │
      ▼
Ejecutar Pruebas
      │
      ▼
Construir Imagen Docker
      │
      ▼
Publicar Imagen
      │
      ▼
Desplegar Proceso ETL
```

Este flujo garantiza que únicamente versiones validadas lleguen a producción.

## Buenas Prácticas

### Automatizar Todo lo Posible

Reducir tareas manuales mejora la consistencia.

### Mantener Pipelines Simples

Facilita mantenimiento y resolución de problemas.

### Versionar Configuraciones

Los archivos del pipeline deben almacenarse en el repositorio.

### Utilizar Secretos Seguros

Nunca almacenar credenciales directamente en el código.

### Monitorear Ejecuciones

Permite detectar problemas rápidamente.

## Consideraciones Finales

Los pipelines CI/CD están compuestos por múltiples elementos que trabajan conjuntamente para automatizar el ciclo de vida del software. Cada componente cumple una función específica que contribuye a mejorar la calidad, seguridad y velocidad de entrega.

Comprender estos elementos es fundamental antes de construir automatizaciones reales. En el siguiente tema se estudiarán las estrategias de versionamiento y automatización utilizadas en proyectos modernos para gestionar cambios, colaboraciones y liberaciones de software.

# Estrategias de Versionamiento y Automatización



En cualquier proyecto de software es fundamental llevar un control adecuado de los cambios realizados a lo largo del tiempo. A medida que las aplicaciones evolucionan, se agregan nuevas funcionalidades, se corrigen errores y se realizan mejoras que deben ser administradas de manera organizada.

Las estrategias de versionamiento permiten identificar claramente el estado de una aplicación en un momento determinado, mientras que las estrategias de automatización facilitan la gestión de cambios, pruebas y despliegues dentro de los pipelines CI/CD.

La combinación de ambas prácticas mejora la trazabilidad, reduce errores y permite mantener procesos de desarrollo más ordenados y escalables.

## ¿Qué es el Versionamiento?

El versionamiento es el proceso mediante el cual se asignan identificadores a distintas versiones de una aplicación o componente.

Estos identificadores permiten:

- Identificar cambios.
- Mantener historial.
- Facilitar despliegues.
- Administrar compatibilidad.
- Recuperar versiones anteriores.

Ejemplo:

```text
v1.0.0
v1.1.0
v1.2.0
v2.0.0
```

Cada versión representa un estado específico del software.

## Importancia del Versionamiento

Sin un esquema de versionamiento adecuado pueden surgir problemas como:

- Dificultad para identificar cambios.
- Despliegues incorrectos.
- Confusión entre equipos.
- Problemas de compatibilidad.

Por ejemplo:

```text
Aplicación Actual
```

no proporciona información útil sobre su estado.

En cambio:

```text
Aplicación v2.3.1
```

permite identificar claramente la versión desplegada.

## Semantic Versioning (SemVer)

El estándar más utilizado actualmente es Semantic Versioning o SemVer.

Formato:

```text
MAJOR.MINOR.PATCH
```

Ejemplo:

```text
2.4.1
```

Interpretación:

| Componente | Valor |
|------------|--------|
| MAJOR | 2 |
| MINOR | 4 |
| PATCH | 1 |

Cada sección tiene un significado específico.

## Cambios MAJOR

Se incrementa cuando existen cambios incompatibles con versiones anteriores.

Ejemplo:

```text
1.0.0
↓
2.0.0
```

Casos comunes:

- Eliminación de funcionalidades.
- Cambios importantes en APIs.
- Modificaciones incompatibles.

## Cambios MINOR

Se incrementa cuando se agregan nuevas funcionalidades compatibles.

Ejemplo:

```text
1.2.0
↓
1.3.0
```

Casos comunes:

- Nuevos módulos.
- Nuevos endpoints.
- Mejoras funcionales.

## Cambios PATCH

Se incrementa para correcciones menores.

Ejemplo:

```text
1.3.1
↓
1.3.2
```

Casos comunes:

- Corrección de errores.
- Ajustes de seguridad.
- Mejoras internas.

## Ejemplos de Evolución

```text
1.0.0
↓
1.0.1
↓
1.0.2
↓
1.1.0
↓
1.2.0
↓
2.0.0
```

Esta secuencia refleja una evolución ordenada del proyecto.

## Etiquetas (Tags)

Las etiquetas permiten asociar versiones específicas dentro del repositorio.

Ejemplo:

```bash
git tag v1.0.0
```

Listar etiquetas:

```bash
git tag
```

Resultado:

```text
v1.0.0
v1.1.0
v2.0.0
```

Los pipelines CI/CD suelen utilizar estas etiquetas para generar versiones oficiales.

## Versionamiento de Imágenes Docker

Las imágenes Docker también deben versionarse adecuadamente.

Incorrecto:

```text
app:latest
```

Correcto:

```text
app:1.0.0
app:1.1.0
app:2.0.0
```

Ejemplo:

```bash
docker build -t app:1.0.0 .
```

Ventajas:

- Mayor control.
- Mejor trazabilidad.
- Facilidad para revertir cambios.

## Estrategias de Ramas

Las ramas permiten desarrollar funcionalidades de forma aislada.

Ejemplo:

```text
main
│
├── feature-login
├── feature-api
└── bugfix-auth
```

Cada rama puede evolucionar independientemente antes de integrarse al proyecto principal.

## Rama Principal

La rama principal suele denominarse:

```text
main
```

o

```text
master
```

Contiene la versión estable del proyecto.

Generalmente:

```text
main
```

es la rama utilizada para producción.

## Ramas de Funcionalidad

Se utilizan para desarrollar nuevas características.

Ejemplo:

```text
feature-reportes
feature-dashboard
feature-alertas
```

Ventajas:

- Aislamiento.
- Menor riesgo.
- Desarrollo paralelo.

## Ramas de Corrección

Utilizadas para solucionar errores específicos.

Ejemplo:

```text
bugfix-login
bugfix-etl
```

Permiten corregir problemas sin afectar el desarrollo principal.

## Flujo Básico de Trabajo

```text
main
 │
 ▼
feature
 │
 ▼
Pull Request
 │
 ▼
Revisión
 │
 ▼
Merge
 │
 ▼
main
```

Este flujo es ampliamente utilizado en proyectos colaborativos.

## Automatización de Versiones

Los pipelines CI/CD pueden generar versiones automáticamente.

Ejemplo conceptual:

```text
Push
 │
 ▼
Build
 │
 ▼
Tag
 │
 ▼
Release
```

Esto reduce errores manuales y mejora la consistencia.

## Automatización Basada en Tags

Muchos pipelines se activan cuando se crea una etiqueta.

Ejemplo:

```bash
git tag v1.2.0
git push origin v1.2.0
```

Resultado:

```text
Pipeline Ejecutado
```

Esto permite automatizar liberaciones oficiales.

## Versionamiento de Artefactos

Los artefactos generados durante el pipeline también deben versionarse.

Ejemplos:

```text
etl-1.0.0.zip
etl-1.1.0.zip

api-2.0.0.jar
```

Esto facilita la administración de versiones históricas.

## Estrategias de Automatización

La automatización busca minimizar la intervención manual en procesos repetitivos.

Ejemplos:

- Pruebas automáticas.
- Construcción automática.
- Generación de artefactos.
- Publicación automática.
- Despliegues automatizados.

## Automatización de Pruebas

Cada vez que se recibe un cambio:

```text
Git Push
      │
      ▼
Pruebas Automáticas
```

Beneficios:

- Detección temprana de errores.
- Mayor calidad.
- Retroalimentación rápida.

## Automatización de Builds

Ejemplo:

```text
Código
   │
   ▼
Build Automático
   │
   ▼
Imagen Docker
```

# Archivos YAML para GitHub Actions

Los workflows de GitHub Actions se definen mediante archivos YAML. Estos archivos contienen todas las instrucciones necesarias para indicar cuándo debe ejecutarse una automatización, qué tareas debe realizar y en qué entorno debe ejecutarse.

Comprender la estructura de estos archivos es fundamental para construir pipelines CI/CD eficientes, ya que representan el núcleo de toda automatización dentro de GitHub Actions.

En este tema se estudiarán los principales componentes de un workflow, su sintaxis y las configuraciones más utilizadas en proyectos reales.

## ¿Qué es YAML?

YAML (YAML Ain't Markup Language) es un formato de serialización de datos diseñado para ser fácil de leer y escribir.

GitHub Actions utiliza YAML para definir workflows debido a su simplicidad y claridad.

Ejemplo:

```yaml
nombre: Aplicación

version: 1.0

entorno:
  desarrollo: true
```

La estructura depende de la indentación para representar jerarquías.

## Reglas Básicas de Sintaxis

### Utilizar Espacios

YAML utiliza espacios para definir niveles jerárquicos.

Correcto:

```yaml
jobs:
  test:
    runs-on: ubuntu-latest
```

Incorrecto:

```yaml
jobs:
test:
runs-on: ubuntu-latest
```

### Mantener Indentación Consistente

Generalmente se utilizan dos espacios por nivel.

### Respetar Mayúsculas y Minúsculas

Las claves son sensibles a mayúsculas y minúsculas.

Correcto:

```yaml
name:
```

Incorrecto:

```yaml
Name:
```

## Ubicación de los Workflows

Todos los workflows deben almacenarse dentro de la siguiente estructura:

```text
.github/
└── workflows/
    └── pipeline.yml
```

Ejemplo:

```text
.github/workflows/ci.yml
```

GitHub detectará automáticamente los archivos ubicados en esta carpeta.

## Estructura General de un Workflow

Un workflow básico contiene los siguientes elementos:

```yaml
name:

on:

jobs:
```

Ejemplo:

```yaml
name: Pipeline CI

on:
  push:

jobs:
  test:
    runs-on: ubuntu-latest
```

Estos componentes representan la base de cualquier automatización.

## Definición del Nombre

La propiedad `name` permite asignar un nombre descriptivo al workflow.

Ejemplo:

```yaml
name: Validación de Proyecto Python
```

Este nombre será visible dentro de la interfaz de GitHub.

## Eventos Disparadores

La sección `on` define cuándo debe ejecutarse el workflow.

Ejemplo:

```yaml
on:
  push:
```

Cada vez que se realice un push al repositorio, el workflow será ejecutado.

## Múltiples Eventos

Es posible definir varios eventos.

Ejemplo:

```yaml
on:
  push:
  pull_request:
```

En este caso el workflow se ejecutará para ambos eventos.

## Filtrado por Rama

Es posible limitar la ejecución a ramas específicas.

Ejemplo:

```yaml
on:
  push:
    branches:
      - main
```

Resultado:

```text
Solo se ejecuta en main
```

## Múltiples Ramas

```yaml
on:
  push:
    branches:
      - main
      - develop
```

El workflow se ejecutará únicamente en esas ramas.

## Ejecución Manual

GitHub permite ejecutar workflows manualmente.

Ejemplo:

```yaml
on:
  workflow_dispatch:
```

Posteriormente podrá ejecutarse desde la interfaz web.

## Ejecución Programada

Los workflows pueden programarse utilizando expresiones cron.

Ejemplo:

```yaml
on:
  schedule:
    - cron: '0 2 * * *'
```

Resultado:

```text
Ejecución diaria a las 2:00 AM
```

Esta funcionalidad es útil para:

- ETLs.
- Reportes.
- Procesos periódicos.

## Definición de Jobs

Los jobs representan grupos de tareas relacionadas.

Ejemplo:

```yaml
jobs:

  test:

  build:
```

Cada job puede ejecutarse de manera independiente.

## Configuración del Runner

Todo job debe indicar el entorno de ejecución.

Ejemplo:

```yaml
runs-on: ubuntu-latest
```

Opciones disponibles:

```yaml
runs-on: windows-latest
```

```yaml
runs-on: macos-latest
```

```yaml
runs-on: ubuntu-latest
```

La opción más utilizada suele ser Ubuntu.

## Definición de Steps

Los steps representan acciones individuales dentro de un job.

Ejemplo:

```yaml
jobs:

  test:

    runs-on: ubuntu-latest

    steps:

      - name: Primer Paso
```

Cada step realiza una tarea específica.

## Ejecución de Comandos

Para ejecutar comandos directamente se utiliza:

```yaml
run:
```

Ejemplo:

```yaml
- name: Mostrar Mensaje
  run: echo "Hola GitHub Actions"
```

Resultado:

```text
Hola GitHub Actions
```

## Ejecución de Múltiples Comandos

```yaml
- name: Comandos
  run: |
    echo "Paso 1"
    echo "Paso 2"
```

Salida:

```text
Paso 1
Paso 2
```

## Uso de Actions

Las acciones reutilizables se incorporan mediante:

```yaml
uses:
```

Ejemplo:

```yaml
- uses: actions/checkout@v4
```

Esta acción descarga automáticamente el código del repositorio.

## Checkout del Repositorio

Prácticamente todos los workflows comienzan con:

```yaml
steps:

  - uses: actions/checkout@v4
```

Esto permite acceder al contenido del repositorio durante la ejecución.

## Configuración de Python

GitHub proporciona una acción oficial para instalar Python.

Ejemplo:

```yaml
- uses: actions/setup-python@v5

  with:
    python-version: "3.12"
```

Resultado:

```text
Python 3.12 disponible
```

## Instalación de Dependencias

Ejemplo:

```yaml
- name: Instalar Dependencias

  run: |
    pip install -r requirements.txt
```

Esto permite preparar el entorno para pruebas o ejecución.

## Variables de Entorno

Las variables pueden definirse mediante:

```yaml
env:
```

Ejemplo:

```yaml
env:
  APP_ENV: production
```

Acceso desde comandos:

```yaml
- run: echo $APP_ENV
```

Resultado:

```text
production
```

## Variables a Nivel de Job

```yaml
jobs:

  test:

    env:
      APP_ENV: testing
```

La variable estará disponible para todos los steps del job.

## Parámetros de Entrada

Algunas actions aceptan parámetros mediante:

```yaml
with:
```

Ejemplo:

```yaml
- uses: actions/setup-python@v5

  with:
    python-version: "3.12"
```

La sección `with` permite personalizar el comportamiento de la acción.

## Dependencias entre Jobs

Los jobs pueden depender unos de otros.

Ejemplo:

```yaml
jobs:

  test:

  build:

    needs: test
```

Resultado:

```text
Build se ejecuta únicamente si Test finaliza correctamente
```

## Artefactos

Los workflows pueden generar archivos para etapas posteriores.

Ejemplos:

- Reportes.
- Resultados.
- Paquetes.
- Archivos comprimidos.

Estos artefactos pueden almacenarse temporalmente dentro de GitHub Actions.

## Workflow Completo para Python

Ejemplo básico:

```yaml
name: Pipeline Python

on:
  push:

jobs:

  test:

    runs-on: ubuntu-latest

    steps:

      - uses: actions/checkout@v4

      - uses: actions/setup-python@v5

        with:
          python-version: "3.12"

      - name: Instalar Dependencias

        run: |
          pip install -r requirements.txt

      - name: Ejecutar Script

        run: |
          python app.py
```

Este workflow:

1. Se ejecuta en cada push.
2. Descarga el repositorio.
3. Instala Python.
4. Instala dependencias.
5. Ejecuta la aplicación.

Beneficios:

Los archivos YAML constituyen el núcleo de GitHub Actions. A través de ellos es posible definir eventos, configurar entornos, ejecutar tareas y automatizar procesos completos de integración y despliegue.

# Construcción Automática de Imágenes Docker

Una vez que una aplicación ha sido validada mediante pruebas automatizadas, el siguiente paso dentro de un pipeline CI/CD consiste en construir la imagen Docker correspondiente. Automatizar este proceso permite garantizar que cada versión aprobada del código genere una imagen consistente y lista para ser distribuida o desplegada.

GitHub Actions permite integrar Docker directamente dentro de los workflows, automatizando la construcción, etiquetado y publicación de imágenes sin intervención manual.

Esta práctica es ampliamente utilizada en proyectos modernos porque facilita la distribución de aplicaciones y asegura que todos los entornos utilicen exactamente la misma versión del software.

## ¿Por Qué Automatizar la Construcción de Imágenes?

Sin automatización, cada nueva versión requeriría ejecutar manualmente:

```bash
docker build -t mi_app .
```

Posteriormente:

```bash
docker push mi_app
```

Este proceso puede generar:

- Errores humanos.
- Versiones inconsistentes.
- Olvido de actualizaciones.
- Retrasos en despliegues.

La automatización elimina estos problemas.

## Flujo General

La construcción automática suele seguir la siguiente secuencia:

```text
Push
 │
 ▼
GitHub Actions
 │
 ▼
Pruebas
 │
 ▼
Docker Build
 │
 ▼
Docker Registry
```

Solo las versiones validadas continúan hacia la etapa de construcción.

## Requisitos Previos

Antes de automatizar la construcción de imágenes es necesario contar con:

- Repositorio GitHub.
- Dockerfile funcional.
- Proyecto validado.
- Cuenta en Docker Hub o un registro compatible.

Estructura básica:

```text
proyecto/
│
├── app.py
├── requirements.txt
├── Dockerfile
└── .github/
    └── workflows/
```

## Dockerfile de Ejemplo

Archivo:

```dockerfile
FROM python:3.12

WORKDIR /app

COPY requirements.txt .

RUN pip install -r requirements.txt

COPY . .

CMD ["python", "app.py"]
```

Este Dockerfile será utilizado por GitHub Actions para construir la imagen.

## Construcción Manual

Recordemos el proceso tradicional:

```bash
docker build -t python-app .
```

Resultado:

```text
python-app:latest
```

GitHub Actions automatizará este procedimiento.

## Acción Oficial de Docker

Docker proporciona acciones oficiales para integrarse con GitHub Actions.

Las más utilizadas son:

```text
docker/setup-buildx-action
docker/login-action
docker/build-push-action
```

Estas acciones simplifican enormemente el proceso de construcción y publicación.

## Configuración Inicial del Workflow

Archivo:

```text
.github/workflows/docker-build.yml
```

Estructura básica:

```yaml
name: Docker Build

on:
  push:

jobs:

  docker:

    runs-on: ubuntu-latest
```

Este workflow se ejecutará en cada push.

## Descarga del Código

Primer paso:

```yaml
steps:

  - uses: actions/checkout@v4
```

Permite acceder al contenido del repositorio.

Docker Buildx proporciona funcionalidades avanzadas de construcción.

Parámetros:

|-----------|-------------|

# Despliegue Automatizado con GitHub Actions



Después de construir y publicar una imagen Docker, el siguiente paso dentro de un pipeline CI/CD consiste en desplegar automáticamente la aplicación en un entorno de ejecución. Esta automatización permite reducir tiempos de entrega, minimizar errores manuales y garantizar que las nuevas versiones lleguen rápidamente a los usuarios.

El despliegue automatizado es una de las etapas más importantes dentro de una estrategia DevOps moderna, ya que conecta directamente el proceso de desarrollo con los entornos de operación.

## ¿Qué es un Despliegue Automatizado?

Un despliegue automatizado consiste en transferir una nueva versión de una aplicación hacia un entorno de ejecución mediante procesos automáticos.

Flujo tradicional:

```text
Desarrollador
      │
      ▼
Construcción
      │
      ▼
Copiar Archivos
      │
      ▼
Actualizar Servidor
```

Flujo automatizado:

```text
Push
 │
 ▼
GitHub Actions
 │
 ▼
Deploy
 │
 ▼
Servidor Actualizado
```

La automatización elimina tareas repetitivas y reduce errores humanos.

## Beneficios del Despliegue Automatizado

### Mayor Velocidad

Las nuevas versiones pueden publicarse en minutos.

### Menor Riesgo

Los procedimientos son repetibles y consistentes.

### Menos Errores Humanos

Se eliminan tareas manuales propensas a fallos.

### Mejor Trazabilidad

Cada despliegue queda registrado.

### Escalabilidad

Permite gestionar múltiples entornos fácilmente.

## Entornos de Despliegue

Normalmente las aplicaciones pasan por varios entornos antes de llegar a producción.

```text
Desarrollo
     │
     ▼
Pruebas
     │
     ▼
Staging
     │
     ▼
Producción
```

Cada entorno cumple una función específica dentro del ciclo de vida del software.

## Estrategia Básica de Despliegue

Un pipeline sencillo puede representarse así:

```text
Push
 │
 ▼
Pruebas
 │
 ▼
Docker Build
 │
 ▼
Docker Registry
 │
 ▼
Deploy
```

Únicamente las versiones que superan todas las validaciones continúan hacia producción.

Esto ayuda a proteger los entornos productivos.

## Dependencia entre Jobs

```text
Test
 │
 ▼
Build
 │
 ▼
Deploy
```

## Mantener Repositorios Organizados

Una estructura clara facilita el mantenimiento del proyecto.

Ejemplo:

```text
proyecto/
│
├── src/
├── tests/
├── Dockerfile
├── requirements.txt
├── compose.yml
└── .github/
    └── workflows/
```

Beneficios:

- Mayor claridad.
- Mejor colaboración.
- Menor curva de aprendizaje.

## Utilizar Control de Versiones Adecuadamente

Todos los cambios deben registrarse mediante Git.

Buenas prácticas:

```text
Commit Pequeños
Commits Frecuentes
Mensajes Claros
```

Ejemplo:

```bash
git commit -m "Agrega validación de usuarios"
```

Evitar:

```bash
git commit -m "Cambios"
```

## Mantener Ramas Organizadas

Ejemplo:

```text
main
│
├── feature-login
├── feature-api
└── bugfix-reportes
```

Beneficios:

- Desarrollo paralelo.
- Menor riesgo.
- Mejor control de cambios.

## Utilizar Semantic Versioning

Versionar aplicaciones utilizando:

```text
MAJOR.MINOR.PATCH
```

Ejemplos:

```text
1.0.0
1.1.0
1.1.1
2.0.0
```

Esto facilita:

- Auditorías.
- Rollbacks.
- Compatibilidad.

## Evitar el Uso Exclusivo de latest

Incorrecto:

```text
python:latest
```

Correcto:

```text
python:3.12
```

o

```text
app:1.4.0
```

Beneficios:

- Mayor estabilidad.
- Resultados reproducibles.
- Menor riesgo de cambios inesperados.

## Optimizar Dockerfiles

Un Dockerfile eficiente reduce tiempos de construcción y tamaño de imagen.

Correcto:

```dockerfile
COPY requirements.txt .

RUN pip install -r requirements.txt

COPY . .
```

Esto aprovecha la caché de Docker.

Beneficios:

- Builds más rápidos.
- Menor consumo de recursos.

## Utilizar Archivos .dockerignore

Ejemplo:

```text
.git
__pycache__
venv
.env
.ipynb_checkpoints
```

Beneficios:

- Imágenes más pequeñas.
- Construcciones más rápidas.
- Mejor seguridad.

## Mantener Imágenes Livianas

Siempre que sea posible utilizar imágenes reducidas.

Ejemplo:

```dockerfile
FROM python:3.12-slim
```

Ventajas:

- Menor espacio.
- Menor tiempo de descarga.
- Menor superficie de ataque.

## No Almacenar Credenciales en el Código

Incorrecto:

```python
PASSWORD = "admin123"
```

Correcto:

```python
import os

PASSWORD = os.getenv("DB_PASSWORD")
```

Toda información sensible debe mantenerse fuera del código fuente.

## Utilizar Secretos en GitHub Actions

Credenciales recomendadas como secretos:

```text
API_KEY
DATABASE_PASSWORD
DOCKER_PASSWORD
AWS_ACCESS_KEY
```

Acceso:

```yaml
${{ secrets.API_KEY }}
```

Nunca deben escribirse directamente en los workflows.

## Automatizar Pruebas

Cada cambio debe validarse automáticamente.

Ejemplo:

```yaml
- run: pytest
```

Beneficios:

- Menos errores.
- Mayor confianza.
- Retroalimentación rápida.

## Validar Código Antes de Construir

Flujo recomendado:

```text
Push
 │
 ▼
Pruebas
 │
 ▼
Build
```

Evita construir imágenes defectuosas.

## Mantener Pipelines Pequeños y Modulares

Incorrecto:

```text
Un único workflow gigante
```

Correcto:

```text
tests.yml
docker.yml
deploy.yml
```

Ventajas:

- Mantenimiento sencillo.
- Mayor reutilización.
- Mejor organización.

## Utilizar Dependencias Entre Jobs

Ejemplo:

```yaml
build:

  needs: test
```

Beneficio:

```text
No construir si las pruebas fallan
```

Esto reduce consumo innecesario de recursos.

# Monitoreo y Solución de Problemas en Docker y GitHub Actions

Una vez que una aplicación ha sido contenerizada y automatizada mediante pipelines CI/CD, es necesario contar con mecanismos que permitan supervisar su funcionamiento y detectar problemas rápidamente.

En entornos reales, una gran parte del trabajo de los equipos DevOps e Ingeniería de Datos consiste en diagnosticar errores relacionados con contenedores, pipelines, despliegues y configuraciones.

Por esta razón, resulta fundamental conocer las herramientas y técnicas básicas de monitoreo y troubleshooting disponibles en Docker y GitHub Actions.

## Importancia del Monitoreo

Aunque una aplicación haya sido desplegada correctamente, esto no garantiza que continuará funcionando adecuadamente.

Pueden ocurrir situaciones como:

- Errores de configuración.
- Consumo excesivo de memoria.
- Fallos de red.
- Problemas de dependencias.
- Errores en pipelines.
- Problemas de autenticación.

Por ello es necesario monitorear continuamente el estado de los sistemas.

## Estrategia General de Diagnóstico

Una metodología sencilla para resolver problemas consiste en seguir el siguiente flujo:

```text
Problema Detectado
        │
        ▼
Identificar Componente
        │
        ▼
Revisar Logs
        │
        ▼
Analizar Error
        │
        ▼
Aplicar Corrección
        │
        ▼
Validar Resultado
```

Este proceso ayuda a evitar diagnósticos apresurados.

## Diagnóstico de Contenedores

El primer paso consiste en verificar si el contenedor está ejecutándose correctamente.

Consultar contenedores activos:

```bash
docker ps
```

Consultar todos los contenedores:

```bash
docker ps -a
```

Ejemplo:

```text
CONTAINER ID   IMAGE      STATUS
abc123         nginx      Up 10 minutes
def456         postgres   Exited
```

Un estado `Exited` indica que el contenedor finalizó su ejecución.

## Consultar Logs de un Contenedor

Los logs son la principal fuente de información para identificar errores.

```bash
docker logs mi_contenedor
```

Ejemplo:

```text
Connection refused
```

o

```text
Database not found
```

Estos mensajes permiten identificar rápidamente la causa del problema.

## Monitoreo en Tiempo Real

Para observar los eventos mientras ocurren:

```bash
docker logs -f mi_contenedor
```

Salida:

```text
Aplicación iniciada...
Conectando a base de datos...
```

La opción `-f` significa follow.

## Inspección de Contenedores

Para obtener información detallada:

```bash
docker inspect mi_contenedor
```

Información disponible:

- Variables de entorno.
- Red.
- Volúmenes.
- Configuración.
- Puertos.

Esta información suele ser útil para detectar configuraciones incorrectas.

## Verificación de Recursos

Consultar consumo de CPU y memoria:

```bash
docker stats
```

Ejemplo:

```text
CONTAINER   CPU %   MEM USAGE
api         2.1%    150MB
postgres    1.3%    420MB
```

Permite detectar:

- Fugas de memoria.
- Procesos excesivos.
- Cuellos de botella.

## Acceso Directo al Contenedor

Para ingresar a un contenedor:

```bash
docker exec -it mi_contenedor bash
```

Una vez dentro:

```bash
ls
pwd
cat archivo.txt
```

Esto facilita la inspección directa del entorno.

## Problemas Frecuentes en Docker

### Puerto No Disponible

Error:

```text
Bind for 0.0.0.0:8080 failed
```

Posible causa:

```text
Puerto ya utilizado
```

Verificar:

```bash
docker ps
```

o

```bash
netstat -ano
```

## Imagen No Encontrada

Error:

```text
Image not found
```

Verificar:

```bash
docker images
```

y confirmar el nombre correcto.

## Volumen No Disponible

Error:

```text
No such volume
```

Consultar:

```bash
docker volume ls
```

## Problemas de Red

Consultar redes:

```bash
docker network ls
```

Inspeccionar:

```bash
docker network inspect nombre_red
```

Permite verificar:

- Contenedores conectados.
- Direcciones IP.
- Configuración DNS.

## Diagnóstico de Workflows

Cuando un pipeline falla, GitHub Actions proporciona registros detallados.

Ruta:

```text
Repositorio
    │
Actions
    │
Workflow
```

Cada ejecución almacena información completa del proceso.

## Estados de Ejecución

GitHub Actions utiliza distintos estados.

```text
Queued
Running
Success
Failed
Cancelled
```

Estos estados permiten identificar rápidamente el resultado del pipeline.

## Revisión de Logs en GitHub Actions

Seleccionar una ejecución.

Ejemplo:

```text
Workflow
    │
Job
    │
Step
```

Cada step posee registros independientes.

Salida típica:

```text
Installing dependencies...
Running tests...
```

Esto facilita localizar exactamente dónde ocurrió el problema.

## Errores Frecuentes en Pipelines

### Dependencias No Instaladas

Error:

```text
ModuleNotFoundError
```

Solución:

Verificar:

```yaml
pip install -r requirements.txt
```

## Error de Sintaxis

Error:

```text
SyntaxError
```

Solución:

Validar localmente:

```bash
python app.py
```

antes de realizar el push.

## Error de Credenciales

Error:

```text
Authentication failed
```

Verificar:

```text
Secrets
```

Configurados en GitHub Actions.

## Error de Docker Login

Error:

```text
unauthorized
```

Verificar:

```text
DOCKER_USERNAME
DOCKER_PASSWORD
```

y confirmar que los secretos sean correctos.

## Monitoreo de Imágenes Docker

Consultar imágenes disponibles:

```bash
docker images
```

Ejemplo:

```text
REPOSITORY     TAG
api            1.0.0
api            latest
```

Permite verificar:

- Versiones.
- Tamaños.
- Fechas.

## Limpieza del Entorno

Eliminar contenedores detenidos:

```bash
docker container prune
```

Eliminar imágenes sin uso:

```bash
docker image prune -a
```

Eliminar volúmenes sin uso:

```bash
docker volume prune
```

Estas tareas ayudan a liberar espacio.

## Estrategia de Resolución de Problemas

Cuando un pipeline falla:

```text
1. Revisar Logs
2. Identificar Step
3. Reproducir Localmente
4. Corregir
5. Ejecutar Nuevamente
```

Cuando un contenedor falla:

```text
1. Verificar Estado
2. Revisar Logs
3. Revisar Configuración
4. Validar Recursos
5. Reiniciar Servicio
```
