# Introducción al Modelado de Datos Analíticos



## ¿Qué es el Modelado de Datos Analíticos?

El modelado de datos analíticos es el proceso de organizar información empresarial de manera estructurada para facilitar:

- análisis de información,
- generación de reportes,
- dashboards,
- indicadores de negocio,
- métricas,
- Business Intelligence,
- y toma de decisiones.

A diferencia de los modelos tradicionales utilizados en sistemas operacionales, el modelado analítico está optimizado para:

- consultas complejas,
- análisis históricos,
- agregaciones,
- lectura masiva de datos,
- y visualización de información.

El objetivo principal es transformar datos en información útil para el negocio.

---

## Diferencia entre Modelado Transaccional y Analítico

| Característica | Modelado Transaccional | Modelado Analítico |
|---|---|---|
| Objetivo | Registrar operaciones | Analizar información |
| Tipo de consultas | Simples | Complejas |
| Operaciones frecuentes | INSERT, UPDATE, DELETE | SELECT |
| Estructura | Altamente normalizada | Optimizada para análisis |
| Uso principal | Sistemas operativos | Business Intelligence |

---

## Arquitectura Analítica Moderna

En una arquitectura moderna de análisis de datos, la información normalmente sigue el siguiente flujo:

```text
Fuentes de Datos
       ↓
Procesos ETL / ELT
       ↓
Data Warehouse
       ↓
Modelo STAR / Snowflake
       ↓
Dashboards y KPIs
```

---

## Flujo de Datos Empresarial

Una empresa puede tener información distribuida en múltiples sistemas:

- ventas,

- recursos humanos,

- logística,

El modelado analítico permite integrar toda esta información para generar análisis globales.

## Ejemplo Práctico Completo

```python
import pandas as pd
import random
from faker import Faker
import matplotlib.pyplot as plt

fake = Faker()

# =========================
# CREACIÓN DE CLIENTES
# =========================

clientes = []

for i in range(1, 11):
    clientes.append({
        "cliente_id": i,
        "nombre": fake.name(),
        "ciudad": fake.city(),
        "edad": random.randint(18, 60)
    })

df_clientes = pd.DataFrame(clientes)

print(df_clientes.head())

# =========================
# CREACIÓN DE PRODUCTOS
# =========================

categorias = [
    "Laptops",
    "Monitores",
    "Mouse",
    "Teclados"
]

productos = []

for i in range(1, 11):
    productos.append({
        "producto_id": i,
        "producto": fake.word(),
        "categoria": random.choice(categorias),
        "precio": round(random.uniform(100, 5000), 2)
    })

df_productos = pd.DataFrame(productos)

print(df_productos.head())

# =========================
# CREACIÓN DE VENTAS
# =========================

ventas = []

for i in range(1, 51):

    cantidad = random.randint(1, 5)

    precio = round(random.uniform(100, 5000), 2)

    ventas.append({
        "venta_id": i,
        "cliente_id": random.randint(1, 10),
        "producto_id": random.randint(1, 10),
        "cantidad": cantidad,
        "total": round(cantidad * precio, 2)
    })

df_ventas = pd.DataFrame(ventas)

print(df_ventas.head())

# =========================
# INTEGRACIÓN DEL MODELO
# =========================

modelo = df_ventas.merge(
    df_clientes,
    on="cliente_id"
).merge(
    df_productos,
    on="producto_id"
)

print(modelo.head())

# =========================
# ANÁLISIS DE VENTAS
# =========================

ventas_categoria = modelo.groupby(
    "categoria"
)["total"].sum()

print(ventas_categoria)

# =========================
# VISUALIZACIÓN
# =========================

ventas_categoria.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Ventas por Categoría")

plt.xlabel("Categoría")

plt.ylabel("Ventas Totales")

plt.show()
```

# Modelado Dimensional



## ¿Qué es el Modelado Dimensional?

El modelado dimensional es una técnica utilizada en Business Intelligence y Data Warehousing para estructurar información de manera optimizada para el análisis de datos.

Su objetivo principal es facilitar:

- consultas rápidas,
- generación de reportes,
- análisis históricos,
- dashboards,
- y métricas empresariales.

Fue popularizado por Ralph Kimball y actualmente sigue siendo uno de los enfoques más utilizados en analítica empresarial.

---

## Características del Modelado Dimensional

El modelado dimensional se caracteriza por:

- simplicidad estructural,
- rapidez en consultas,
- facilidad de comprensión,
- orientación al negocio,
- y optimización para análisis.

A diferencia de modelos altamente normalizados, el modelo dimensional prioriza el rendimiento analítico.

---

## Componentes Principales

Un modelo dimensional se compone principalmente de:

- tablas fact,
- tablas dimensión.

---

## Tabla Fact

Las tablas fact contienen:

- métricas,
- indicadores,
- datos cuantitativos,
- transacciones.

Normalmente almacenan grandes volúmenes de registros.

---

## Ejemplos de métricas

- ventas,
- costos,
- utilidades,
- salarios,
- impuestos,
- cantidades vendidas.

---

## Ejemplo conceptual

```text
FACT_VENTAS
--------------------------------
venta_id
cliente_id
producto_id
fecha_id
sucursal_id
cantidad
precio
total
```

---

## Tabla Dimensión

Las tablas dimensión contienen información descriptiva utilizada para analizar las métricas.

---

## Ejemplos de dimensiones

- clientes,
- productos,
- empleados,
- tiempo,
- sucursales,
- departamentos.

---

## Ejemplo conceptual

```text
DIM_PRODUCTOS
-------------------------
producto_id
nombre_producto
categoria
marca
precio
```

---

## Diferencia entre Fact y Dimensión

| Tabla Fact | Tabla Dimensión |
|---|---|
| Contiene métricas | Contiene descripciones |
| Gran volumen | Menor volumen |
| Valores numéricos | Valores textuales |
| Cambia constantemente | Cambia menos |
| Base del análisis | Contexto del análisis |

---

## Granularidad

La granularidad define el nivel de detalle de la información almacenada en la tabla fact.

Es una de las decisiones más importantes del modelado dimensional.

---

## Ejemplo

### Granularidad diaria

```text
1 fila = 1 venta por día
```

### Granularidad por transacción

```text
1 fila = 1 producto vendido
```

Mientras más fina sea la granularidad:
- mayor detalle,
- mayor tamaño de almacenamiento.

---

## Jerarquías

Las jerarquías permiten analizar información en distintos niveles.

---

## Ejemplo de jerarquía temporal

```text
Año
 ↓
Mes
 ↓
Día
```

---

## Ejemplo de jerarquía geográfica

```text
País
 ↓
Departamento
 ↓
Ciudad
 ↓
Sucursal
```

---

## Dimensiones Conformadas

Son dimensiones compartidas entre múltiples procesos analíticos.

---

## Ejemplo

La dimensión tiempo puede ser utilizada en:

- ventas,
- recursos humanos,
- inventarios,
- logística.

# Ejemplo Práctico Completo

```python
import pandas as pd
import random
from faker import Faker

fake = Faker()

# =====================================
# CREACIÓN DE DIMENSIÓN CLIENTES
# =====================================

clientes = []

for i in range(1, 11):

    clientes.append({
        "cliente_id": i,
        "nombre_cliente": fake.name(),
        "ciudad": fake.city(),
        "edad": random.randint(18, 65)
    })

dim_clientes = pd.DataFrame(clientes)

print(dim_clientes.head())

# =====================================
# CREACIÓN DE DIMENSIÓN PRODUCTOS
# =====================================

categorias = [
    "Laptops",
    "Monitores",
    "Mouse",
    "Teclados"
]

productos = []

for i in range(1, 11):

    productos.append({
        "producto_id": i,
        "producto": fake.word(),
        "categoria": random.choice(categorias),
        "precio": round(random.uniform(200, 5000), 2)
    })

dim_productos = pd.DataFrame(productos)

print(dim_productos.head())

# =====================================
# CREACIÓN DE DIMENSIÓN TIEMPO
# =====================================

fechas = []

for i in range(1, 31):

    fechas.append({
        "fecha_id": i,
        "dia": i,
        "mes": "Mayo",
        "anio": 2026
    })

dim_tiempo = pd.DataFrame(fechas)

print(dim_tiempo.head())

# =====================================
# CREACIÓN DE TABLA FACT
# =====================================

ventas = []

for i in range(1, 101):

    cantidad = random.randint(1, 5)

    precio = round(random.uniform(200, 5000), 2)

    ventas.append({
        "venta_id": i,
        "cliente_id": random.randint(1, 10),
        "producto_id": random.randint(1, 10),
        "fecha_id": random.randint(1, 30),
        "cantidad": cantidad,
        "total": round(cantidad * precio, 2)
    })

fact_ventas = pd.DataFrame(ventas)

print(fact_ventas.head())

# =====================================
# INTEGRACIÓN DEL MODELO
# =====================================

modelo = fact_ventas.merge(
    dim_clientes,
    on="cliente_id"
).merge(
    dim_productos,
    on="producto_id"
).merge(
    dim_tiempo,
    on="fecha_id"
)

print(modelo.head())

# =====================================
# ANÁLISIS ANALÍTICO
# =====================================

ventas_categoria = modelo.groupby(
    "categoria"
)["total"].sum()

print(ventas_categoria)

# =====================================
# ANÁLISIS POR CIUDAD
# =====================================

ventas_ciudad = modelo.groupby(
    "ciudad"
)["total"].sum()

print(ventas_ciudad)

# =====================================
# TOP 5 CLIENTES
# =====================================

top_clientes = modelo.groupby(
    "nombre_cliente"
)["total"].sum().sort_values(
    ascending=False
).head(5)

print(top_clientes)
```

 # Diseño de Tablas Fact y Dimension



## Introducción

Uno de los pasos más importantes en el modelado dimensional es el diseño correcto de:

- tablas fact,
- tablas dimensión.

Estas estructuras serán la base para:
- dashboards,
- reportes,
- KPIs,
- análisis históricos,
- y procesos de Business Intelligence.

Un diseño incorrecto puede generar:
- bajo rendimiento,
- duplicidad de información,
- errores analíticos,
- y dificultad para escalar el sistema.

---

## Tablas Fact

Las tablas fact almacenan:

- métricas,
- indicadores,
- transacciones,
- eventos del negocio.

Generalmente contienen:
- claves foráneas,
- datos numéricos,
- y grandes volúmenes de información.

---

### Características de una Tabla Fact

| Característica | Descripción |
|---|---|
| Contiene métricas | Sí |
| Alto volumen de registros | Sí |
| Datos numéricos | Principalmente |
| Cambios frecuentes | Sí |
| Uso analítico | Principal |

---

### Ejemplo de Tabla Fact

```text
FACT_VENTAS
--------------------------------
venta_id
cliente_id
producto_id
fecha_id
sucursal_id
cantidad
precio
total
```

---

## Métricas Más Comunes

### Métricas Aditivas

Pueden sumarse completamente.

Ejemplos:
- ventas,
- cantidades,
- ingresos.

---

### Métricas Semi-Aditivas

Pueden agregarse parcialmente.

Ejemplo:
- saldos bancarios.

---

### Métricas No Aditivas

No pueden sumarse.

Ejemplo:
- porcentajes,
- promedios.

---

## Tablas Dimensión

Las tablas dimensión contienen información descriptiva utilizada para contextualizar las métricas.

---

### Características de una Tabla Dimensión

| Característica | Descripción |
|---|---|
| Información descriptiva | Sí |
| Menor volumen | Sí |
| Datos textuales | Principalmente |
| Cambios menos frecuentes | Sí |
| Facilitan filtros | Sí |

---

### Ejemplo de Tabla Dimensión

```text
DIM_CLIENTES
--------------------------
cliente_id
nombre
ciudad
edad
genero
```

---

## Tipos de Dimensiones

### Dimensiones Simples

Contienen atributos descriptivos básicos.

Ejemplo:

```text
DIM_PRODUCTOS
```

---

### Dimensiones Conformadas

Son compartidas por varios procesos analíticos.

Ejemplo:

```text
DIM_TIEMPO
```

Puede ser utilizada por:
- ventas,
- recursos humanos,
- inventarios,
- logística.

---

### Dimensiones Degeneradas

Son identificadores almacenados directamente en la tabla fact.

Ejemplo:

```text
numero_factura
numero_transaccion
```

---

### Junk Dimensions

Agrupan múltiples atributos pequeños de baja cardinalidad.

Ejemplo:

```text
estado_pago
tipo_cliente
estado_envio
```

---

## Claves Sustitutas (Surrogate Keys)

Las claves sustitutas son identificadores artificiales utilizados en dimensiones.

---

### Ejemplo

```text
cliente_key
```

en lugar de:

```text
dpi_cliente
```

---

### Ventajas de las Claves Sustitutas

- Mejor rendimiento.
- Independencia del sistema origen.
- Facilitan historización.
- Simplifican relaciones.
- Evitan inconsistencias.

---

## Relaciones entre Fact y Dimensiones

```text
DIM_CLIENTES ----|
                 |
DIM_PRODUCTOS ---|---- FACT_VENTAS
                 |
DIM_TIEMPO ------|
```

---

## Granularidad en la Tabla Fact

La granularidad define el nivel exacto de detalle de cada fila.

---

### Ejemplo

#### Granularidad por transacción

```text
1 fila = 1 producto vendido
```

#### Granularidad diaria

```text
1 fila = ventas del día
```

---

### Importancia de la Granularidad

Una mala granularidad puede:
- duplicar métricas,
- afectar rendimiento,
- dificultar análisis.

---

## Diseño Correcto de una Fact Table

Una tabla fact ideal debe:

- tener granularidad clara,
- contener métricas relevantes,
- incluir claves foráneas,
- evitar atributos descriptivos innecesarios.

---

## Diseño Correcto de una Dimension Table

Una dimensión ideal debe:

- contener atributos descriptivos,
- ser entendible,
- facilitar filtros,
- permitir agrupaciones.

# STAR Schema (Esquema Estrella)



## Introducción

El STAR Schema o Esquema Estrella es uno de los modelos más utilizados en Business Intelligence y Data Warehousing.

Su objetivo es organizar información de forma optimizada para:

- consultas analíticas,
- dashboards,
- KPIs,
- reportes,
- y análisis históricos.

Recibe el nombre de “estrella” debido a su estructura visual:

- una tabla fact en el centro,
- múltiples dimensiones alrededor.

---

## Estructura del STAR Schema

Un STAR Schema se compone de:

- una tabla fact central,
- varias tablas dimensión conectadas directamente.

---

## Estructura Conceptual

```text
                DIM_CLIENTES
                       |
                       |
DIM_PRODUCTOS --- FACT_VENTAS --- DIM_TIEMPO
                       |
                       |
                DIM_SUCURSALES
```

---

## Tabla Fact en STAR Schema

La tabla fact contiene:

- métricas,
- indicadores,
- transacciones,
- eventos del negocio.

---

### Ejemplo

```text
FACT_VENTAS
--------------------------------
venta_id
cliente_key
producto_key
fecha_key
sucursal_key
cantidad
precio
descuento
total
```

---

## Tablas Dimensión en STAR Schema

Las dimensiones contienen información descriptiva.

---

### Ejemplo de Dimensiones

```text
DIM_CLIENTES
```

```text
DIM_PRODUCTOS
```

```text
DIM_TIEMPO
```

```text
DIM_SUCURSALES
```

---

## Características del STAR Schema

| Característica | Descripción |
|---|---|
| Modelo simple | Sí |
| Fácil de entender | Sí |
| Alto rendimiento analítico | Sí |
| Dimensiones desnormalizadas | Sí |
| Consultas rápidas | Sí |

---

## Ventajas del STAR Schema

- Consultas rápidas.
- Fácil mantenimiento.
- Fácil comprensión.
- Compatible con herramientas BI.
- Ideal para dashboards.
- Menor complejidad SQL.

---

## Desventajas del STAR Schema

- Puede existir redundancia.
- Mayor espacio de almacenamiento.
- Dimensiones grandes pueden crecer rápidamente.

---

## ¿Por qué se usan Dimensiones Desnormalizadas?

En STAR Schema las dimensiones normalmente NO se normalizan completamente.

Esto se hace para:

- mejorar rendimiento,
- reducir JOINs,
- simplificar consultas.

---

## Ejemplo de Dimensión Desnormalizada

```text
DIM_PRODUCTOS
--------------------------------
producto_key
producto
categoria
marca
proveedor
pais_origen
```

Toda la información se mantiene en una sola tabla.

---

## Flujo Analítico en STAR Schema

```text
Fuentes de Datos
       ↓
ETL / ELT
       ↓
Tabla Fact
       ↓
Dimensiones
       ↓
Modelo STAR
       ↓
Dashboard
```

# Snowflake Schema (Esquema Copo de Nieve)



## Introducción

El Snowflake Schema o Esquema Copo de Nieve es una variante del STAR Schema utilizada en Data Warehousing y Business Intelligence.

Su principal característica es que las dimensiones se encuentran normalizadas, lo que genera una estructura más compleja pero con menor redundancia de información.

El nombre “Snowflake” proviene de la forma visual que toma el modelo, similar a un copo de nieve.

---

## Estructura del Snowflake Schema

En un Snowflake Schema:

- la tabla fact permanece en el centro,
- las dimensiones se dividen en múltiples tablas relacionadas.

---

## Estructura Conceptual

```text
                    DIM_CIUDADES
                           |
                           |
DIM_CLIENTES ----- DIM_REGIONES
        |
        |
        |------ FACT_VENTAS ------ DIM_PRODUCTOS ----- DIM_CATEGORIAS
        |
        |
   DIM_TIEMPO
```

---

## Diferencia Principal con STAR Schema

| STAR Schema | Snowflake Schema |
|---|---|
| Dimensiones desnormalizadas | Dimensiones normalizadas |
| Menos JOINs | Más JOINs |
| Más simple | Más complejo |
| Más redundancia | Menos redundancia |
| Mayor velocidad | Menor velocidad relativa |

---

## ¿Qué significa Normalizar Dimensiones?

Normalizar significa dividir información en varias tablas relacionadas para reducir redundancia.

---

## Ejemplo

### STAR Schema

```text
DIM_PRODUCTOS
--------------------------------
producto_key
producto
categoria
marca
proveedor
pais_origen
```

Toda la información se almacena en una sola tabla.

---

### Snowflake Schema

```text
DIM_PRODUCTOS
-------------------------
producto_key
producto
categoria_key
marca_key
```

```text
DIM_CATEGORIAS
-------------------------
categoria_key
categoria
```

```text
DIM_MARCAS
-------------------------
marca_key
marca
pais_origen
```

---

## Ventajas del Snowflake Schema

- Reduce redundancia.
- Menor espacio de almacenamiento.
- Mayor integridad de datos.
- Facilita mantenimiento de dimensiones grandes.
- Mejor organización estructural.

---

## Desventajas del Snowflake Schema

- Consultas más complejas.
- Mayor cantidad de JOINs.
- Menor rendimiento analítico comparado con STAR.
- Más difícil de entender para usuarios no técnicos.

---

## ¿Cuándo se Utiliza Snowflake Schema?

El Snowflake Schema suele utilizarse cuando:

- existen dimensiones muy grandes,
- se requiere reducir almacenamiento,
- se necesita alta integridad,
- o las dimensiones tienen muchas jerarquías.

---

## Ejemplo Empresarial

### Retail

Separación de:

- categorías,
- marcas,
- proveedores,
- regiones,
- departamentos.

---

### Bancos

Separación de:

- productos financieros,
- tipos de cuenta,
- segmentos de clientes,
- regiones.

---

## Flujo Analítico

```text
Fuentes de Datos
       ↓
ETL / ELT
       ↓
Normalización de Dimensiones
       ↓
Snowflake Schema
       ↓
Dashboards
```

## Buenas Prácticas

- Normalizar únicamente cuando sea necesario.
- Evitar exceso de tablas pequeñas.

## Casos Ideales para Snowflake

- empresas con grandes volúmenes,

# STAR vs Snowflake Schema



## Introducción

El STAR Schema y el Snowflake Schema son los dos modelos dimensionales más utilizados en Business Intelligence y Data Warehousing.

Ambos tienen el mismo objetivo:

- facilitar análisis,
- optimizar consultas,
- construir dashboards,
- generar KPIs,
- y organizar información analítica.

Sin embargo, poseen diferencias importantes en:

- estructura,
- rendimiento,
- complejidad,
- almacenamiento,
- y mantenimiento.

---

## STAR Schema

El STAR Schema utiliza:

- una tabla fact central,
- dimensiones desnormalizadas.

---

### Estructura STAR

```text
                DIM_CLIENTES
                       |
                       |
DIM_PRODUCTOS --- FACT_VENTAS --- DIM_TIEMPO
                       |
                       |
                DIM_SUCURSALES
```

---

### Características del STAR Schema

- Modelo simple.
- Menos JOINs.
- Consultas rápidas.
- Fácil mantenimiento.
- Dimensiones grandes y desnormalizadas.

---

## Snowflake Schema

El Snowflake Schema utiliza:

- una tabla fact central,
- dimensiones normalizadas.

---

### Estructura Snowflake

```text
                   DIM_CATEGORIAS
                           |
                           |
DIM_PRODUCTOS ----- FACT_VENTAS ----- DIM_TIEMPO
       |
       |
DIM_MARCAS
```

---

### Características del Snowflake Schema

- Modelo más complejo.
- Más JOINs.
- Menor redundancia.
- Mejor integridad estructural.
- Dimensiones normalizadas.

---

## Diferencia Principal

La diferencia principal es:

### STAR Schema

Las dimensiones permanecen desnormalizadas.

---

### Snowflake Schema

Las dimensiones se dividen en múltiples tablas relacionadas.

---

## Comparación General

| Característica | STAR Schema | Snowflake Schema |
|---|---|---|
| Complejidad | Baja | Alta |
| Rendimiento | Alto | Medio |
| JOINs | Menos | Más |
| Redundancia | Mayor | Menor |
| Espacio de almacenamiento | Mayor | Menor |
| Facilidad de análisis | Alta | Media |
| Facilidad de mantenimiento | Alta | Media |
| Normalización | Baja | Alta |

---

## Comparación de Dimensiones

### STAR Schema

```text
DIM_PRODUCTOS
--------------------------------
producto_key
producto
categoria
marca
proveedor
pais_origen
```

Toda la información se encuentra en una sola tabla.

---

### Snowflake Schema

```text
DIM_PRODUCTOS
-------------------------
producto_key
producto
categoria_key
marca_key
```

```text
DIM_CATEGORIAS
-------------------------
categoria_key
categoria
```

```text
DIM_MARCAS
-------------------------
marca_key
marca
pais_origen
```

La información se divide en múltiples dimensiones.

---

## Comparación de Rendimiento

### STAR Schema

Generalmente posee mejor rendimiento porque:
- requiere menos JOINs,
- las consultas son más simples,
- y la lectura es más rápida.

---

### Snowflake Schema

Puede tener menor rendimiento debido a:
- múltiples JOINs,
- dimensiones fragmentadas,
- y consultas más complejas.

---

## Comparación de Almacenamiento

### STAR Schema

- mayor redundancia,
- más espacio utilizado.

---

### Snowflake Schema

- menor redundancia,
- menor almacenamiento requerido.

---

## Comparación de Mantenimiento

### STAR Schema

Más fácil de mantener para:
- analistas,
- usuarios BI,
- equipos pequeños.

---

### Snowflake Schema

Más adecuado para:
- arquitecturas corporativas,
- grandes organizaciones,
- modelos complejos.

---

## Casos Ideales para STAR Schema

- dashboards rápidos,
- análisis comerciales,
- modelos simples,
- proyectos BI pequeños y medianos.

---

## Casos Ideales para Snowflake Schema

- grandes corporaciones,
- dimensiones complejas,
- jerarquías extensas,
- reducción de redundancia,
- integración empresarial avanzada.

---

## Ejemplo SQL STAR Schema

## Ejemplo Práctico Completo

```python
import pandas as pd
import random
from faker import Faker
import time

fake = Faker()

# ==========================================
# CREACIÓN DE DIMENSIONES STAR
# ==========================================

productos_star = []

categorias = [
    "Laptops",
    "Monitores",
    "Mouse",
    "Teclados"
]

marcas = [
    "HP",
    "Dell",
    "Lenovo",
    "Logitech"
]

for i in range(1, 101):

    productos_star.append({
        "producto_key": i,
        "producto": fake.word(),
        "categoria": random.choice(categorias),
        "marca": random.choice(marcas),
        "pais_origen": random.choice([
            "USA",
            "China",
            "México"
        ])
    })

dim_productos_star = pd.DataFrame(productos_star)

# ==========================================
# CREACIÓN DE DIMENSIONES SNOWFLAKE
# ==========================================

dim_categorias = pd.DataFrame({
    "categoria_key": [1,2,3,4],
    "categoria": categorias
})

dim_marcas = pd.DataFrame({
    "marca_key": [1,2,3,4],
    "marca": marcas
})

productos_snowflake = []

for i in range(1, 101):

    productos_snowflake.append({
        "producto_key": i,
        "producto": fake.word(),
        "categoria_key": random.randint(1,4),
        "marca_key": random.randint(1,4)
    })

dim_productos_snowflake = pd.DataFrame(
    productos_snowflake
)

# ==========================================
# FACT TABLE
# ==========================================

ventas = []

for i in range(1, 50001):

    ventas.append({
        "venta_id": i,
        "producto_key": random.randint(1,100),
        "total": round(
            random.uniform(100, 5000),
            2
        )
    })

fact_ventas = pd.DataFrame(ventas)

# ==========================================
# PRUEBA STAR SCHEMA
# ==========================================

inicio_star = time.time()

modelo_star = fact_ventas.merge(
    dim_productos_star,
    on="producto_key"
)

ventas_star = modelo_star.groupby(
    "categoria"
)["total"].sum()

fin_star = time.time()

tiempo_star = fin_star - inicio_star

print("Resultado STAR")
print(ventas_star)

print("\nTiempo STAR:")
print(tiempo_star)

# ==========================================
# PRUEBA SNOWFLAKE SCHEMA
# ==========================================

inicio_snowflake = time.time()

modelo_snowflake = fact_ventas.merge(
    dim_productos_snowflake,
    on="producto_key"
).merge(
    dim_categorias,
    on="categoria_key"
).merge(
    dim_marcas,
    on="marca_key"
)

ventas_snowflake = modelo_snowflake.groupby(
    "categoria"
)["total"].sum()

fin_snowflake = time.time()

tiempo_snowflake = (
    fin_snowflake - inicio_snowflake
)

print("\nResultado Snowflake")
print(ventas_snowflake)

print("\nTiempo Snowflake:")
print(tiempo_snowflake)

# ==========================================
# COMPARACIÓN FINAL
# ==========================================

print("\nComparación de Rendimiento")

if tiempo_star < tiempo_snowflake:

    print(
        "STAR Schema fue más rápido"
    )

else:

    print(
        "Snowflake Schema fue más rápido"
    )
```

# Slowly Changing Dimensions (SCD)



## Introducción

En los sistemas analíticos y Data Warehouses, uno de los problemas más comunes es el manejo de cambios históricos en las dimensiones.

Por ejemplo:

- un cliente cambia de ciudad,
- un empleado cambia de departamento,
- un producto cambia de categoría,
- una sucursal cambia de región.

Las Slowly Changing Dimensions (SCD) permiten administrar estos cambios sin perder consistencia analítica.

---

## ¿Qué son las Slowly Changing Dimensions?

Las Slowly Changing Dimensions son técnicas utilizadas para manejar cambios en los atributos de las tablas dimensión a lo largo del tiempo.

El objetivo es decidir:

- si se sobrescriben datos,
- si se conserva historial,
- o si se mantiene una combinación de ambos.

---

## Importancia de las SCD

Las SCD son fundamentales porque permiten:

- análisis históricos,
- auditoría,
- trazabilidad,
- análisis temporal,
- y consistencia en KPIs históricos.

---

## Problema Empresarial

Supongamos el siguiente caso:

```text
Cliente: Juan Pérez
Ciudad: Guatemala
```

Meses después:

```text
Cliente: Juan Pérez
Ciudad: Antigua Guatemala
```

La pregunta es:

### ¿Qué debe ocurrir con el historial anterior?

---

## Tipos de Slowly Changing Dimensions

Los tipos más utilizados son:

- SCD Tipo 1,
- SCD Tipo 2,
- SCD Tipo 3.

---

## SCD Tipo 1

En el SCD Tipo 1:

- el dato anterior se sobrescribe,
- no se conserva historial.

---

### Funcionamiento

```text
ANTES

cliente_key | nombre       | ciudad
-----------------------------------
1           | Juan Pérez   | Guatemala
```

```text
DESPUÉS

cliente_key | nombre       | ciudad
-----------------------------------
1           | Juan Pérez   | Antigua Guatemala
```

La ciudad anterior desaparece.

---

### Ventajas

- Simplicidad.
- Menor almacenamiento.
- Fácil mantenimiento.

---

### Desventajas

- Se pierde historial.
- No permite auditoría histórica.

---

### Casos Ideales

- errores de escritura,
- correcciones simples,
- atributos no históricos.

---

## SCD Tipo 2

En el SCD Tipo 2:

- se conserva historial completo,
- se crea un nuevo registro.

Es el tipo más utilizado en Data Warehousing.

---

### Funcionamiento

```text
ANTES

cliente_key | nombre       | ciudad
-----------------------------------
1           | Juan Pérez   | Guatemala
```

```text
DESPUÉS

cliente_key | nombre       | ciudad
-----------------------------------
1           | Juan Pérez   | Guatemala

2           | Juan Pérez   | Antigua Guatemala
```

---

## Nuevos Campos Utilizados

El SCD Tipo 2 normalmente agrega:

```text
fecha_inicio
fecha_fin
registro_activo
```

---

### Ejemplo

```text
cliente_key | nombre | ciudad             | activo
---------------------------------------------------
1           | Juan   | Guatemala          | No

2           | Juan   | Antigua Guatemala  | Sí
```

---

### Ventajas

- Conserva historial completo.
- Permite análisis históricos.
- Ideal para auditoría.

---

### Desventajas

- Mayor almacenamiento.
- Mayor complejidad.

---

### Casos Ideales

- análisis históricos,
- recursos humanos,
- bancos,
- retail,
- comportamiento de clientes.

---

## SCD Tipo 3

En el SCD Tipo 3:

- únicamente se conserva un cambio anterior.

---

### Funcionamiento

```text
cliente_key | nombre | ciudad_actual | ciudad_anterior
-------------------------------------------------------
1           | Juan   | Antigua       | Guatemala
```

---

### Ventajas

- Conserva historial limitado.
- Fácil implementación.

---

### Desventajas

- Historial parcial.
- Poco flexible.

---

### Casos Ideales

- cambios pequeños,
- comparación simple de estados.

---

## Comparación de Tipos SCD

| Tipo | Historial | Complejidad | Uso |
|---|---|---|---|
| Tipo 1 | No | Baja | Correcciones |
| Tipo 2 | Sí | Alta | Auditoría e histórico |
| Tipo 3 | Parcial | Media | Comparaciones simples |

---

```python
import pandas as pd
from datetime import datetime

# ==========================================
# CREACIÓN DE DIMENSIÓN INICIAL
# ==========================================

dim_clientes = pd.DataFrame({

    "cliente_key": [1],

    "nombre": ["Juan Pérez"],

    "ciudad": ["Guatemala"],

    "fecha_inicio": ["2026-01-01"],

    "fecha_fin": [None],

    "activo": ["Sí"]

})

print("DIMENSIÓN ORIGINAL")

print(dim_clientes)

# ==========================================
# SCD TIPO 1
# ==========================================

print("\n==============================")
print("SCD TIPO 1")
print("==============================")

dim_tipo1 = dim_clientes.copy()

dim_tipo1.loc[
    dim_tipo1["cliente_key"] == 1,
    "ciudad"
] = "Antigua Guatemala"

print(dim_tipo1)

# ==========================================
# SCD TIPO 2
# ==========================================

print("\n==============================")
print("SCD TIPO 2")
print("==============================")

dim_tipo2 = dim_clientes.copy()

# CERRAR REGISTRO ANTERIOR

dim_tipo2.loc[
    dim_tipo2["cliente_key"] == 1,
    "fecha_fin"
] = "2026-06-01"

dim_tipo2.loc[
    dim_tipo2["cliente_key"] == 1,
    "activo"
] = "No"

# NUEVO REGISTRO

nuevo_registro = pd.DataFrame({

    "cliente_key": [2],

    "nombre": ["Juan Pérez"],

    "ciudad": ["Antigua Guatemala"],

    "fecha_inicio": ["2026-06-01"],

    "fecha_fin": [None],

    "activo": ["Sí"]

})

dim_tipo2 = pd.concat([
    dim_tipo2,
    nuevo_registro
])

print(dim_tipo2)

# ==========================================
# SCD TIPO 3
# ==========================================

print("\n==============================")
print("SCD TIPO 3")
print("==============================")

dim_tipo3 = pd.DataFrame({

    "cliente_key": [1],

    "nombre": ["Juan Pérez"],

    "ciudad_actual": ["Guatemala"],

    "ciudad_anterior": [None]

})

# ACTUALIZACIÓN

dim_tipo3["ciudad_anterior"] = (
    dim_tipo3["ciudad_actual"]
)

dim_tipo3["ciudad_actual"] = (
    "Antigua Guatemala"
)

print(dim_tipo3)

# ==========================================
# COMPARACIÓN FINAL
# ==========================================

print("\n==============================")
print("COMPARACIÓN FINAL")
print("==============================")

print("\nTIPO 1:")
print(dim_tipo1)

print("\nTIPO 2:")
print(dim_tipo2)

print("\nTIPO 3:")
print(dim_tipo3)
```

# Procesos ETL y ELT para Modelos Analíticos



## Introducción

Los procesos ETL y ELT son fundamentales en Data Engineering, Business Intelligence y Data Warehousing.

Permiten:

- integrar datos,
- limpiar información,
- transformar estructuras,
- automatizar procesos analíticos,
- y alimentar modelos STAR y Snowflake.

Sin ETL o ELT, un Data Warehouse no podría mantenerse actualizado correctamente.

---

## ¿Qué es ETL?

ETL significa:

```text
Extract
Transform
Load
```

Es un proceso donde:

1. se extraen datos,
2. se transforman,
3. y luego se cargan al Data Warehouse.

---

## Flujo ETL

```text
Fuentes de Datos
       ↓
EXTRACT
       ↓
TRANSFORM
       ↓
LOAD
       ↓
Data Warehouse
```

---

## ¿Qué es ELT?

ELT significa:

```text
Extract
Load
Transform
```

En este modelo:

1. los datos se extraen,
2. se cargan primero,
3. y luego se transforman dentro del Data Warehouse.

---

## Flujo ELT

```text
Fuentes de Datos
       ↓
EXTRACT
       ↓
LOAD
       ↓
Data Warehouse
       ↓
TRANSFORM
```

---

## Diferencia entre ETL y ELT

| Característica | ETL | ELT |
|---|---|---|
| Transformación | Antes de cargar | Después de cargar |
| Rendimiento | Medio | Alto |
| Uso moderno | Menor | Mayor |
| Big Data | Limitado | Mejor |
| Complejidad | Media | Alta |

---

## Ejemplo Práctico Completo

```python
import pandas as pd
import random
from faker import Faker

fake = Faker()

# ==========================================
# EXTRACT
# ==========================================

print("FASE EXTRACT")

clientes = []

for i in range(1, 21):

    clientes.append({
        "cliente_id": i,
        "nombre": fake.name(),
        "ciudad": fake.city(),
        "edad": random.randint(18, 70)
    })

df_clientes = pd.DataFrame(clientes)

print(df_clientes.head())

# ==========================================
# EXTRACT VENTAS
# ==========================================

ventas = []

for i in range(1, 101):

    cantidad = random.randint(1, 10)

    precio = round(
        random.uniform(100, 5000),
        2
    )

    ventas.append({
        "venta_id": i,
        "cliente_id": random.randint(1, 20),
        "cantidad": cantidad,
        "precio": precio
    })

df_ventas = pd.DataFrame(ventas)

print(df_ventas.head())

# ==========================================
# TRANSFORM
# ==========================================

print("\nFASE TRANSFORM")

# CREACIÓN DE TOTAL

df_ventas["total"] = (
    df_ventas["cantidad"] *
    df_ventas["precio"]
)

# VALIDACIÓN DE NULOS

print("\nNULOS:")

print(df_ventas.isnull().sum())

# ELIMINAR DUPLICADOS

df_ventas = df_ventas.drop_duplicates()

# CONVERSIÓN DE TIPOS

df_ventas["cantidad"] = (
    df_ventas["cantidad"].astype(int)
)

print(df_ventas.head())

# ==========================================
# INTEGRACIÓN
# ==========================================

modelo = df_ventas.merge(
    df_clientes,
    on="cliente_id"
)

print("\nMODELO INTEGRADO")

print(modelo.head())

# ==========================================
# LOAD
# ==========================================

print("\nFASE LOAD")

# SIMULACIÓN DE CARGA

fact_ventas = modelo.copy()

print("\nFACT TABLE CREADA")

print(fact_ventas.head())

# ==========================================
# KPIs
# ==========================================

print("\nVENTAS POR CIUDAD")

ventas_ciudad = modelo.groupby(
    "ciudad"
)["total"].sum()

print(ventas_ciudad)

# ==========================================
# KPI CLIENTES
# ==========================================

print("\nTOP CLIENTES")

top_clientes = modelo.groupby(
    "nombre"
)["total"].sum().sort_values(
    ascending=False
).head(10)

print(top_clientes)

# ==========================================
# KPI PROMEDIO
# ==========================================

print("\nPROMEDIO DE VENTA")

print(modelo["total"].mean())
```

## Ejemplo Práctico Completo

```python
import pandas as pd
import random
from faker import Faker
from sqlalchemy import create_engine

fake = Faker()

# ==========================================
# CREACIÓN DE CONEXIÓN SQLITE
# ==========================================

engine = create_engine(
    "sqlite:///datawarehouse.db"
)

print("Conexión exitosa")

# ==========================================
# CREACIÓN DE DIM_CLIENTES
# ==========================================

clientes = []

for i in range(1, 21):

    clientes.append({
        "cliente_key": i,
        "cliente": fake.name(),
        "ciudad": fake.city()
    })

dim_clientes = pd.DataFrame(clientes)

print(dim_clientes.head())

# ==========================================
# CREACIÓN DE DIM_PRODUCTOS
# ==========================================

categorias = [
    "Laptops",
    "Monitores",
    "Mouse",
    "Teclados"
]

productos = []

for i in range(1, 16):

    productos.append({
        "producto_key": i,
        "producto": fake.word(),
        "categoria": random.choice(categorias)
    })

dim_productos = pd.DataFrame(productos)

print(dim_productos.head())

# ==========================================
# CREACIÓN DE FACT_VENTAS
# ==========================================

ventas = []

for i in range(1, 501):

    cantidad = random.randint(1, 10)

    precio = round(
        random.uniform(100, 5000),
        2
    )

    total = cantidad * precio

    ventas.append({
        "venta_id": i,
        "cliente_key": random.randint(1, 20),
        "producto_key": random.randint(1, 15),
        "cantidad": cantidad,
        "precio": precio,
        "total": round(total, 2)
    })

fact_ventas = pd.DataFrame(ventas)

print(fact_ventas.head())

# ==========================================
# CARGA A BASE DE DATOS
# ==========================================

dim_clientes.to_sql(
    "dim_clientes",
    engine,
    if_exists="replace",
    index=False
)

dim_productos.to_sql(
    "dim_productos",
    engine,
    if_exists="replace",
    index=False
)

fact_ventas.to_sql(
    "fact_ventas",
    engine,
    if_exists="replace",
    index=False
)

print("Datos cargados correctamente")

# ==========================================
# LECTURA DESDE SQL
# ==========================================

clientes_sql = pd.read_sql(
    "SELECT * FROM dim_clientes",
    engine
)

productos_sql = pd.read_sql(
    "SELECT * FROM dim_productos",
    engine
)

ventas_sql = pd.read_sql(
    "SELECT * FROM fact_ventas",
    engine
)

print(clientes_sql.head())

# ==========================================
# CONSTRUCCIÓN DEL MODELO
# ==========================================

modelo = ventas_sql.merge(
    clientes_sql,
    on="cliente_key"
).merge(
    productos_sql,
    on="producto_key"
)

print(modelo.head())

# ==========================================
# KPI 1 - VENTAS POR CATEGORÍA
# ==========================================

ventas_categoria = modelo.groupby(
    "categoria"
)["total"].sum()

print("\nVENTAS POR CATEGORÍA")

print(ventas_categoria)

# ==========================================
# KPI 2 - VENTAS POR CIUDAD
# ==========================================

ventas_ciudad = modelo.groupby(
    "ciudad"
)["total"].sum()

print("\nVENTAS POR CIUDAD")

print(ventas_ciudad)

# ==========================================
# KPI 3 - TOP CLIENTES
# ==========================================

top_clientes = modelo.groupby(
    "cliente"
)["total"].sum().sort_values(
    ascending=False
).head(10)

print("\nTOP CLIENTES")

print(top_clientes)

# ==========================================
# KPI 4 - PROMEDIO DE VENTA
# ==========================================

promedio = modelo["total"].mean()

print("\nPROMEDIO DE VENTA")

print(promedio)
```

```python
import pandas as pd
import random
from faker import Faker
from sqlalchemy import create_engine

fake = Faker()

# ==========================================
# CREACIÓN DE CONEXIÓN SQLITE
# ==========================================

engine = create_engine(
    "sqlite:///datawarehouse.db"
)

print("Conexión exitosa")

# ==========================================
# CREACIÓN DE DIM_CLIENTES
# ==========================================

clientes = []

for i in range(1, 21):

    clientes.append({
        "cliente_key": i,
        "cliente": fake.name(),
        "ciudad": fake.city()
    })

dim_clientes = pd.DataFrame(clientes)

print(dim_clientes.head())

# ==========================================
# CREACIÓN DE DIM_PRODUCTOS
# ==========================================

categorias = [
    "Laptops",
    "Monitores",
    "Mouse",
    "Teclados"
]

productos = []

for i in range(1, 16):

    productos.append({
        "producto_key": i,
        "producto": fake.word(),
        "categoria": random.choice(categorias)
    })

dim_productos = pd.DataFrame(productos)

print(dim_productos.head())

# ==========================================
# CREACIÓN DE FACT_VENTAS
# ==========================================

ventas = []

for i in range(1, 501):

    cantidad = random.randint(1, 10)

    precio = round(
        random.uniform(100, 5000),
        2
    )

    total = cantidad * precio

    ventas.append({
        "venta_id": i,
        "cliente_key": random.randint(1, 20),
        "producto_key": random.randint(1, 15),
        "cantidad": cantidad,
        "precio": precio,
        "total": round(total, 2)
    })

fact_ventas = pd.DataFrame(ventas)

print(fact_ventas.head())

# ==========================================
# CARGA A BASE DE DATOS
# ==========================================

dim_clientes.to_sql(
    "dim_clientes",
    engine,
    if_exists="replace",
    index=False
)

dim_productos.to_sql(
    "dim_productos",
    engine,
    if_exists="replace",
    index=False
)

fact_ventas.to_sql(
    "fact_ventas",
    engine,
    if_exists="replace",
    index=False
)

print("Datos cargados correctamente")

# ==========================================
# LECTURA DESDE SQL
# ==========================================

clientes_sql = pd.read_sql(
    "SELECT * FROM dim_clientes",
    engine
)

productos_sql = pd.read_sql(
    "SELECT * FROM dim_productos",
    engine
)

ventas_sql = pd.read_sql(
    "SELECT * FROM fact_ventas",
    engine
)

print(clientes_sql.head())

# ==========================================
# CONSTRUCCIÓN DEL MODELO
# ==========================================

modelo = ventas_sql.merge(
    clientes_sql,
    on="cliente_key"
).merge(
    productos_sql,
    on="producto_key"
)

print(modelo.head())

# ==========================================
# KPI 1 - VENTAS POR CATEGORÍA
# ==========================================

ventas_categoria = modelo.groupby(
    "categoria"
)["total"].sum()

print("\nVENTAS POR CATEGORÍA")

print(ventas_categoria)

# ==========================================
# KPI 2 - VENTAS POR CIUDAD
# ==========================================

ventas_ciudad = modelo.groupby(
    "ciudad"
)["total"].sum()

print("\nVENTAS POR CIUDAD")

print(ventas_ciudad)

# ==========================================
# KPI 3 - TOP CLIENTES
# ==========================================

top_clientes = modelo.groupby(
    "cliente"
)["total"].sum().sort_values(
    ascending=False
).head(10)

print("\nTOP CLIENTES")

print(top_clientes)

# ==========================================
# KPI 4 - PROMEDIO DE VENTA
# ==========================================

promedio = modelo["total"].mean()

print("\nPROMEDIO DE VENTA")

print(promedio)
```

# Consultas Analíticas para Business Intelligence

## Introducción

Las consultas analíticas son fundamentales en Business Intelligence y Data Warehousing.

Permiten:

- generar KPIs,
- construir dashboards,
- analizar comportamiento empresarial,
- detectar patrones,
- y apoyar la toma de decisiones.

A diferencia de consultas transaccionales, las consultas analíticas están enfocadas en:

- agregaciones,
- históricos,
- tendencias,
- rankings,
- comparaciones,
- y métricas empresariales.

---

## Diferencia entre Consultas OLTP y OLAP

| Característica | OLTP | OLAP |
|---|---|---|
| Objetivo | Registrar operaciones | Analizar información |
| Tipo de consulta | Simple | Compleja |
| Uso principal | Sistemas operativos | BI y análisis |
| Operaciones comunes | INSERT, UPDATE | SELECT |
| Volumen analizado | Bajo | Alto |

## Agregaciones SQL

Las agregaciones son la base de la analítica.

---

## SUM()

Calcula sumatorias.

### Ejemplo

```sql
SELECT
    SUM(total) AS ventas_totales
FROM fact_ventas;
```

---

## AVG()

Calcula promedios.

### Ejemplo

```sql
SELECT
    AVG(total) AS promedio_ventas
FROM fact_ventas;
```

---

## COUNT()

Cuenta registros.

### Ejemplo

```sql
SELECT
    COUNT(*) AS cantidad_ventas
FROM fact_ventas;
```

---

## MAX() y MIN()

Obtienen valores máximos y mínimos.

### Ejemplo

```sql
SELECT
    MAX(total) AS venta_maxima,
    MIN(total) AS venta_minima
FROM fact_ventas;
```

---

## GROUP BY

Permite agrupar información.

---

## Ejemplo

### Ventas por categoría

```sql
SELECT
    categoria,
    SUM(total) AS ventas
FROM fact_ventas fv

JOIN dim_productos dp
    ON fv.producto_key = dp.producto_key

GROUP BY categoria;
```

---

## ORDER BY

Permite ordenar resultados.

---

## Ejemplo

```sql
SELECT
    categoria,
    SUM(total) AS ventas
FROM fact_ventas fv

JOIN dim_productos dp
    ON fv.producto_key = dp.producto_key

GROUP BY categoria

ORDER BY ventas DESC;
```

---

## HAVING

Permite filtrar agregaciones.

---

## Ejemplo

```sql
SELECT
    categoria,
    SUM(total) AS ventas
FROM fact_ventas fv

JOIN dim_productos dp
    ON fv.producto_key = dp.producto_key

GROUP BY categoria

HAVING SUM(total) > 100000;
```

---

## Window Functions

Las Window Functions permiten realizar cálculos avanzados sin perder granularidad.

---

## ROW_NUMBER()

Genera numeración.

### Ejemplo

```sql
SELECT
    cliente,
    total,

    ROW_NUMBER() OVER(
        ORDER BY total DESC
    ) AS ranking

FROM fact_ventas;
```

---

## RANK()

Genera rankings.

### Ejemplo

```sql
SELECT
    cliente,
    total,

    RANK() OVER(
        ORDER BY total DESC
    ) AS ranking

FROM fact_ventas;
```

---

## DENSE_RANK()

Similar a RANK pero sin saltos.

---

## Ejemplo

```sql
SELECT
    cliente,
    total,

    DENSE_RANK() OVER(
        ORDER BY total DESC
    ) AS ranking

FROM fact_ventas;
```

---

## SUM OVER()

Permite acumulados.

### Ejemplo

```sql
SELECT
    fecha,
    total,

    SUM(total) OVER(
        ORDER BY fecha
    ) AS acumulado

FROM fact_ventas;
```

## Ejemplo Práctico Completo

```python
import pandas as pd
import random
from faker import Faker

fake = Faker()

# ==========================================
# CREACIÓN DE DIMENSIONES
# ==========================================

clientes = []

for i in range(1, 21):

    clientes.append({
        "cliente_key": i,
        "cliente": fake.name(),
        "ciudad": fake.city()
    })

dim_clientes = pd.DataFrame(clientes)

# ==========================================
# DIM_PRODUCTOS
# ==========================================

categorias = [
    "Laptops",
    "Monitores",
    "Mouse",
    "Teclados"
]

productos = []

for i in range(1, 16):

    productos.append({
        "producto_key": i,
        "producto": fake.word(),
        "categoria": random.choice(categorias)
    })

dim_productos = pd.DataFrame(productos)

# ==========================================
# FACT_VENTAS
# ==========================================

ventas = []

for i in range(1, 1001):

    cantidad = random.randint(1, 10)

    precio = round(
        random.uniform(100, 5000),
        2
    )

    total = cantidad * precio

    ventas.append({
        "venta_id": i,
        "cliente_key": random.randint(1, 20),
        "producto_key": random.randint(1, 15),
        "cantidad": cantidad,
        "precio": precio,
        "total": round(total, 2)
    })

fact_ventas = pd.DataFrame(ventas)

# ==========================================
# MODELO ANALÍTICO
# ==========================================

modelo = fact_ventas.merge(
    dim_clientes,
    on="cliente_key"
).merge(
    dim_productos,
    on="producto_key"
)

print(modelo.head())

# ==========================================
# KPI 1 - VENTAS TOTALES
# ==========================================

ventas_totales = modelo["total"].sum()

print("\nVENTAS TOTALES")

print(ventas_totales)

# ==========================================
# KPI 2 - PROMEDIO DE VENTA
# ==========================================

promedio_ventas = modelo["total"].mean()

print("\nPROMEDIO DE VENTA")

print(promedio_ventas)

# ==========================================
# KPI 3 - VENTAS POR CATEGORÍA
# ==========================================

ventas_categoria = modelo.groupby(
    "categoria"
)["total"].sum()

print("\nVENTAS POR CATEGORÍA")

print(ventas_categoria)

# ==========================================
# KPI 4 - TOP CLIENTES
# ==========================================

top_clientes = modelo.groupby(
    "cliente"
)["total"].sum().sort_values(
    ascending=False
).head(10)

print("\nTOP CLIENTES")

print(top_clientes)

# ==========================================
# KPI 5 - VENTAS POR CIUDAD
# ==========================================

ventas_ciudad = modelo.groupby(
    "ciudad"
)["total"].sum()

print("\nVENTAS POR CIUDAD")

print(ventas_ciudad)

# ==========================================
# KPI 6 - ACUMULADO
# ==========================================

modelo["acumulado"] = modelo[
    "total"
].cumsum()

print("\nACUMULADO")

print(modelo[
    ["venta_id", "total", "acumulado"]
].head())

# ==========================================
# KPI 7 - RANKING
# ==========================================

ranking = modelo.groupby(
    "cliente"
)["total"].sum().rank(
    ascending=False
)

print("\nRANKING CLIENTES")

print(ranking)

# ==========================================
# KPI 8 - VENTA MÁXIMA
# ==========================================

venta_maxima = modelo["total"].max()

print("\nVENTA MÁXIMA")

print(venta_maxima)

# ==========================================
# KPI 9 - VENTA MÍNIMA
# ==========================================

venta_minima = modelo["total"].min()

print("\nVENTA MÍNIMA")

print(venta_minima)
```

# Optimización de Modelos Analíticos



## Introducción

En Business Intelligence y Data Warehousing, la optimización es fundamental para garantizar:

- consultas rápidas,
- dashboards eficientes,
- menor consumo de recursos,
- escalabilidad,
- y mejor experiencia analítica.

Un modelo analítico mal optimizado puede provocar:

- consultas lentas,
- alto consumo de memoria,
- bloqueos,
- duplicidad,
- y KPIs incorrectos.

---

## ¿Qué es la Optimización Analítica?

La optimización analítica consiste en mejorar:

- estructura de datos,
- consultas,
- almacenamiento,
- índices,
- y rendimiento general del sistema analítico.

---

## Objetivos de la Optimización

- reducir tiempos de respuesta,
- minimizar consumo de recursos,
- mejorar rendimiento de JOINs,
- optimizar dashboards,
- y facilitar escalabilidad.

---

## Factores que Afectan el Rendimiento

| Factor | Impacto |
|---|---|
| Índices | Alto |
| JOINs excesivos | Alto |
| Datos duplicados | Alto |
| SELECT * | Medio |
| Falta de filtros | Alto |
| Mala granularidad | Alto |

---

```python
import pandas as pd
import random
import time
from faker import Faker

fake = Faker()

# ==========================================
# CREACIÓN DE DATOS
# ==========================================

ventas = []

for i in range(1, 100001):

    ventas.append({
        "venta_id": i,
        "cliente_key": random.randint(1, 1000),
        "producto_key": random.randint(1, 500),
        "cantidad": random.randint(1, 10),
        "precio": round(
            random.uniform(100, 5000),
            2
        )
    })

fact_ventas = pd.DataFrame(ventas)

# ==========================================
# CREACIÓN DE TOTAL
# ==========================================

fact_ventas["total"] = (
    fact_ventas["cantidad"] *
    fact_ventas["precio"]
)

print(fact_ventas.head())

# ==========================================
# MEDICIÓN DE TIEMPO
# ==========================================

inicio = time.time()

ventas_cliente = fact_ventas.groupby(
    "cliente_key"
)["total"].sum()

fin = time.time()

print("\nVENTAS POR CLIENTE")

print(ventas_cliente.head())

print("\nTIEMPO DE EJECUCIÓN")

print(fin - inicio)

# ==========================================
# OPTIMIZACIÓN DE TIPOS
# ==========================================

fact_ventas["cliente_key"] = (
    fact_ventas["cliente_key"].astype("int32")
)

fact_ventas["producto_key"] = (
    fact_ventas["producto_key"].astype("int32")
)

fact_ventas["cantidad"] = (
    fact_ventas["cantidad"].astype("int16")
)

# ==========================================
# MEMORIA UTILIZADA
# ==========================================

print("\nMEMORIA UTILIZADA")

print(
    fact_ventas.info(
        memory_usage="deep"
    )
)

# ==========================================
# FILTRO OPTIMIZADO
# ==========================================

ventas_filtradas = fact_ventas[
    fact_ventas["total"] > 10000
]

print("\nVENTAS FILTRADAS")

print(ventas_filtradas.head())

# ==========================================
# KPI OPTIMIZADO
# ==========================================

top_clientes = fact_ventas.groupby(
    "cliente_key"
)["total"].sum().sort_values(
    ascending=False
).head(10)

print("\nTOP CLIENTES")

print(top_clientes)

# ==========================================
# AGRUPACIÓN OPTIMIZADA
# ==========================================

ventas_producto = fact_ventas.groupby(
    "producto_key"
)["total"].agg([
    "sum",
    "mean",
    "max"
])

print("\nVENTAS POR PRODUCTO")

print(ventas_producto.head())

# ==========================================
# VALIDACIÓN DE DUPLICADOS
# ==========================================

duplicados = fact_ventas.duplicated().sum()

print("\nDUPLICADOS")

print(duplicados)
```

# Visualización y Validación del Modelo Analítico



## Introducción

En Data Warehousing y Business Intelligence no basta únicamente con construir modelos analíticos.

También es necesario:

- validar relaciones,
- revisar consistencia,
- documentar estructuras,
- visualizar modelos,
- y verificar calidad de datos.

La visualización y validación permiten garantizar que el modelo:

- sea entendible,
- funcione correctamente,
- produzca KPIs confiables,
- y pueda escalar correctamente.

---

## ¿Qué es la Validación del Modelo?

La validación consiste en verificar que:

- las relaciones sean correctas,
- no existan duplicados,
- las métricas sean consistentes,
- la granularidad sea adecuada,
- y los datos sean confiables.

## Validación de Integridad Referencial

### Ejemplo SQL

```sql
SELECT *
FROM fact_ventas fv

LEFT JOIN dim_productos dp
    ON fv.producto_key = dp.producto_key

WHERE dp.producto_key IS NULL;
```

Esto permite detectar claves huérfanas.

## Validación de Duplicados

### SQL

```sql
SELECT
    cliente_key,
    COUNT(*)
FROM dim_clientes

GROUP BY cliente_key

HAVING COUNT(*) > 1;
```

## Ejemplo Práctico Completo

```python
import pandas as pd
import random
from faker import Faker
from graphviz import Digraph

fake = Faker()

# ==========================================
# CREACIÓN DE DIMENSIONES
# ==========================================

clientes = []

for i in range(1, 21):

    clientes.append({
        "cliente_key": i,
        "cliente": fake.name(),
        "ciudad": fake.city()
    })

dim_clientes = pd.DataFrame(clientes)

# ==========================================
# DIM_PRODUCTOS
# ==========================================

productos = []

for i in range(1, 16):

    productos.append({
        "producto_key": i,
        "producto": fake.word(),
        "categoria": random.choice([
            "Laptops",
            "Monitores",
            "Mouse",
            "Teclados"
        ])
    })

dim_productos = pd.DataFrame(productos)

# ==========================================
# FACT_VENTAS
# ==========================================

ventas = []

for i in range(1, 501):

    ventas.append({
        "venta_id": i,
        "cliente_key": random.randint(1, 20),
        "producto_key": random.randint(1, 15),
        "total": round(
            random.uniform(100, 5000),
            2
        )
    })

fact_ventas = pd.DataFrame(ventas)

# ==========================================
# VALIDACIÓN DE DUPLICADOS
# ==========================================

duplicados = fact_ventas.duplicated().sum()

print("\nDUPLICADOS")

print(duplicados)

# ==========================================
# VALIDACIÓN DE NULOS
# ==========================================

print("\nNULOS")

print(
    fact_ventas.isnull().sum()
)

# ==========================================
# CONSTRUCCIÓN DEL MODELO
# ==========================================

modelo = fact_ventas.merge(
    dim_clientes,
    on="cliente_key",
    how="left"
).merge(
    dim_productos,
    on="producto_key",
    how="left"
)

print(modelo.head())

# ==========================================
# VALIDACIÓN DE CLAVES HUÉRFANAS
# ==========================================

huerfanos = modelo[
    modelo["producto"].isnull()
]

print("\nCLAVES HUÉRFANAS")

print(huerfanos)

# ==========================================
# VALIDACIÓN DE KPIs
# ==========================================

ventas_totales = modelo["total"].sum()

print("\nVENTAS TOTALES")

print(ventas_totales)

# ==========================================
# VALIDACIÓN DE CARDINALIDAD
# ==========================================

cardinalidad = fact_ventas.groupby(
    "producto_key"
).size()

print("\nCARDINALIDAD")

print(cardinalidad.head())

# ==========================================
# VISUALIZACIÓN DEL MODELO
# ==========================================

dot = Digraph()

dot.node("FACT_VENTAS")

dot.node("DIM_CLIENTES")

dot.node("DIM_PRODUCTOS")

dot.edge(
    "FACT_VENTAS",
    "DIM_CLIENTES"
)

dot.edge(
    "FACT_VENTAS",
    "DIM_PRODUCTOS"
)

dot.render(
    "modelo_analitico",
    format="png",
    cleanup=True
)

print("\nDIAGRAMA GENERADO")

# ==========================================
# VALIDACIÓN DE TIPOS
# ==========================================

print("\nTIPOS DE DATOS")

print(modelo.dtypes)

# ==========================================
# VALIDACIÓN ESTADÍSTICA
# ==========================================

print("\nESTADÍSTICAS")

print(modelo.describe())
```
